## Agentic AI бенчмарки (2024-2026)  


>DeepSeek-V4-Flash-0731 - удивительно как DeepSeek удалось улучшить свою первоначальную версию модели за счет дополнительного посттренинга, в некоторых задачах результаты выросли в несколько раз!

#### Terminal Bench 2.1 (Terminal-Bench)

Способность ИИ выступать в роли `системного администратора` или инженера автоматизации. Модель запускают в изолированном Docker-контейнере с полноценным терминалом Linux. Ей дают 89 реальных сложных задач: настроить сервер, исправить конфигурацию Git, откомпилировать проект, разобраться с пакетами или обучить модель. (Stanford)

#### NL2Repo (NL2Repo-Bench)

Навык «генерации репозитория с нуля» (Long-Horizon Repository Generation). ИИ дают один текстовый файл с ТЗ и абсолютно пустую рабочую папку. Модель должна сама `спроектировать архитектуру` приложения, прописать зависимости, создать многомодульную логику и на выходе выдать полностью рабочий и устанавливаемый Python-пакет. Тест невероятно длинный и сложный. Например справится с созданием игры с нуля: правильно связать скрипты игрока, физики, интерфейса (UI) и противников в единую структуру, не запутавшись в файлах. (arXiv)

#### Cybergym

Способности ИИ в сфере `информационной безопасности` и этичного хакинга (кибербезопасности). Модель закидывают на полигон, где она должна выявлять уязвимости, проводить тесты на проникновение (пентест) или защищать систему от атак. (ai labs)

#### DeepSWE (SWE-bench)

Задача ИИ — зайти в реальный open-source репозиторий на GitHub, прочитать чужой код, найти сложный баг, о котором сообщает пользователь, и создать рабочий `Pull Request` с исправлением, который пройдет все юнит-тесты. Например если игра выдает ошибку в консоли Unity или Unreal Engine, разобраться в каком файле и строчке сломалась логика. (Princeton+)

#### Toolathlon-Verified (The Tool Decathlon)

Умение ИИ пользоваться внешними инструментами через `API` (календари, мессенджеры, базы данных, веб-браузеры) для решения комплексных бытовых и офисных задач. Версия Verified означает, что траектории выполнения задач проверялись людьми вручную, чтобы исключить ложные срабатывания. (HKUST NLP)

#### Agents' Last Exam / AutomationBench Public

Комплексную `автоматизацию веб-интерфейсов и операционных систем`. Поиск информации на сайтах, перенос данных в таблицы, вызов скриптов, доводить бизнес-процесс до автоматизма без участия оператора. (MLCommons)

#### DSBench (FullStack / Hard (Math) )

Навыки автономного `Data Science`. Содержит сотни реальных задач по анализу и моделированию данных, взятых из реальных соревнований Kaggle. ИИ должен проводить `end-to-end` анализ: загрузить огромные базы данных, построить математические модели, обучить нейросеть и выдать результат. Версия Hard содержит задачи с запутанной логикой и грязными данными. (Academic DS). Версия FullStack - например, генерация процедурного мира, система инвентаря или ИИ для врагов.

## Почему они заслуживают доверия

- Защита от «зубрёжки» (Contamination): Старые тесты ИИ часто сдавали хорошо просто потому, что их вопросы случайно попадали в обучающую выборку из интернета. В таких тестах, как Terminal-Bench или Toolathlon, сжульничать невозможно: ИИ **оценивают** по финальному **результату выполнения работы** (прошел ли код юнит-тесты, решилась ли системная проблема).  

- Оценка в динамической среде: Модели тестируются в интерактивном режиме с **ограничением по времени** и числу попыток. Это приближено к работе реального программиста или аналитика.  

- Авторитетные создатели: За тестами стоят не сами создатели моделей (OpenAI или DeepSeek), а независимые университеты (Stanford, HKUST, Princeton), которые регулярно обновляют версии, чтобы закрывать лазейки для накрутки результатов.

## Какие есть нюансы (почему слепо верить цифрам нельзя)  

- Проблема стоимости и скорости: Некоторые модели получают высокий балл в таблице (например, Opus-4.8), потому что они тратят **огромные вычислительные ресурсы**, делают сотни внутренних циклов размышлений («режим рассуждения») и работают медленно. В реальной работе это может стоить очень дорого.  

- Узкая специализация: Если модель выбилась в лидеры в DSBench-Hard, это гарантирует, что она великолепна в математике и Python. Но это никак не отражает её способность, например, вести приятный диалог с клиентом на человеческом языке или писать художественные тексты.

---

MoE (Mixture of Experts) — это архитектура нейросетей, которая разделяет модель на специализированные части («эксперты») и подключает только нужные из них для обработки каждого конкретного слова или токена.  

В отличие от традиционных (плотных) моделей, где при каждом запросе работают 100% параметров, в MoE-модели активна лишь малая часть сети. Это позволяет создавать гигантские по объему знаний модели, которые работают так же быстро, как маленькие.  

>Поскольку диспетчер (маршрутизатор) заранее не знает, какой именно из `сотен экспертов` модели понадобится для следующего слова, все эти `304 миллиарда` параметров должны одновременно находиться в оперативной памяти видеокарт (VRAM) -> Поэтому потребление памяти рассчитывается на все 304B.

Примеры таких моделей:


- Mixtral 8x7B / 8x22B
- GPT-4
- DeepSeek-V4-Flash-0731

### DeepSeek-V4-Flash-0731  


Параметры:  

- 284B — основная MoE-модель  

- 304B — checkpoint с модулем DSpark

- 13B - активных  (6-8 экспертов)

Контекст:  

- до 1 млн токенов
  
Модули:  

- **DSpark** технология **спекулятивного декодирования** (Speculative Decoding)

1. Внутри этого модуля работает маленькая и очень быстрая «черновая» модель-помощник (Draft Model).
2. Помощник предугадывает сразу несколько следующих токенов вперед с высокой скоростью.
3. Основная большая MoE-модель (на 284B) проверяет эти догадки за один проход.
4. В результате скорость генерации текста (инференс) увеличивается на **60–85%** без потери качества ответов.

---

`Активные параметры:` 13B  

- 5B неизменных слоев (внимание, общие базовые знания)  
- 8B (обьем экспертов).  

Это количество математических операций, которое видеокарта совершает для генерации **одного единственного слова (токена)**. 

- Когда отправляется запрос, встроенный роутер выбирает из сотен доступных экспертов всего **6 самых подходящих** для конкретного слова.  

- Энергия процессоров и графических ядер тратится на просчет только этих 6 выбранных экспертов. 
- **Итог:** Модель думает и генерирует текст со скоростью маленькой, сверхбыстрой **13-миллиардной модели**, но при этом обладает колоссальной эрудицией и базой знаний, накопленной во всех **284–304 миллиардах параметров**.

#### Алгоритм Top-K в MoE моделях (experts router)

1. **Оценка пригодности:** На вход роутеру подается токен (слово). Роутер делает быстрое умножение матриц и выставляет каждому эксперту в модели «оценку» от 0 до 1, показывающую, насколько этот эксперт подходит для решения текущей задачи.  

2. **Фильтрация (Top-K):** Вместо того чтобы отправлять токен всем экспертам, алгоритм сортирует оценки по убыванию и оставляет только **K** лучших. Например, в архитектурах DeepSeek используется схема, где из сотен мелких экспертов выбираются **Top-6** или **Top-8** лучших для данного токена.  

3. **Нормализация (Softmax):** Веса выбранных \(K\) экспертов нормализуются, чтобы в сумме давать 1. Токен отправляется на вычисление параллельно только в эти выбранные слои, а их ответы затем складываются с учетом этих весов.

### Можно ли расширить модель и сделать её ещё умнее?

**Да, можно, и это стандартная практика в современной AI-индустрии.** Сделать MoE-модель «умнее» и больше можно тремя основными путями:

---

1. Изменение шага маршрутизации (Увеличение K) — Самый простой путь

Вы можете заставить роутер выбирать не Top-6 экспертов, а, например, **Top-12** или вообще активировать всех экспертов сразу.

- **Что произойдет:** Модель станет точнее улавливать нюансы, так как в работе будет участвовать больше «мнений» экспертов. Число активных параметров вырастет с 13B до 25B–30B.
- **Проблема:** Скорость генерации (инференса) катастрофически упадет, а вычислительные затраты вырастут пропорционально. Модель потеряет свое главное преимущество — экономичность.

---

2. Метод MoEification (Превращение плоской модели в MoE)

Если у вас есть обычная плотная модель (например, Llama 3 70B), её можно «разрезать» на экспертов. Матрицы весов разделяются на кластеры, и над ними надстраивается роутер. Из 70B плотной модели можно получить MoE-модель на ~200B параметров с 15B активными. Она сохранит умственные способности базовой модели, но начнет работать в разы быстрее.

Как определяются эксперты при «нарезке» плоской модели?

Разрезать модель наугад нельзя — тогда эксперты превратятся в кашу. Чтобы MoEфикация сработала, нейроны группируют по смыслам с помощью **алгоритмов кластеризации** (чаще всего используется метод **K-Means / К-средних**).

Процесс нарезки выглядит так:

1. **Прогон данных (Profiling):** Через исходную плотную модель (например, Llama 3 8B) прогоняют большой массив разнообразных текстов. В этот момент специальный скрипт записывает, какие именно нейроны активируются вместе при появлении определенных тем (например, какие нейроны реагируют на медицинские термины, какие — на код, какие — на знаки препинания).
2. **Кластеризация (Clustering):** Алгоритм К-средних анализирует эти записи и группирует нейроны, которые «дружат» друг с другом и активируются одновременно, в отдельные кластеры.
3. **Формирование экспертов:** Каждый такой кластер физически обособляется и объявляется «Экспертом №X». Например, 100 нейронов, которые всегда активировались на химических формулах, теперь упаковываются в Эксперта №1.
4. **Обучение роутера:** Так как веса экспертов мы просто скопировали из готовой модели, они уже умеют работать. Единственное, что нужно обучить с нуля — это сам роутер. Его учат правильно угадывать, какому кластеру принадлежит пришедшее слово. Это обучение происходит очень быстро (буквально за несколько часов) на обычном текстовом датасете.

MoEфикация — это сложный исследовательский процесс. Она применяется, чтобы ускорить гигантские модели (например, превратить плотную модель 70B в быструю MoE 70B c 15B активными).

Например для задачи (медицина на русском языке) на базе модели 8B в MoEфикации нет смысла: модель 8B и так работает очень быстро, а после нарезки она не станет умнее, она просто станет потреблять чуть меньше вычислительной мощности процессора при том же объеме VRAM.

---

3. FrankenMoE или MergeKit (Слияние разных моделей) — Самый популярный народный метод

С помощью открытых утилит (например, библиотеки `mergekit`) разработчики берут **несколько совершенно разных готовых моделей** одного размера (например, одну модель, обученную на программирование, вторую — на медицину, третью — на художественный текст) и объединяют их под общим роутером.

- Вы буквально создаете «комитет экспертов» своими руками. Из четырех моделей по 8B параметров вы можете собрать одну MoE-модель формата **4x8B** (около 32B общих параметров).
- Такая модель станет значительно умнее во всех сферах сразу, не требуя колоссальных бюджетов на обучение с нуля.
