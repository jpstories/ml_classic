`ndarray` (N-dimensional array) - основной объект для хранения и манипуляций многомерными массивами данных **одного** типа.
- одномерный (вектор)  
- двумерный (матрица)  
- трехмерный  
- и тд  

Можно создать через: np.array(), np.zeros(), np.ones(), np.arange() и другие

---

```python
import numpy as np

# метод, создает массив с числами (с дробими работает не очень)
x_range_int = np.arange(start=0, stop=10, step=1)

# start - Начальное значение последовательности. default=0
# stop - Конечное значение последовательности (не включается)
# step - Шаг между значениями. default=1
# dtype - авто

print(x_range_int)
```

```python
import numpy as np

# метод, создает массив с дробями
x_range_float = np.linspace(start=-1.0, stop=1.0, num=21, endpoint=True, retstep=False)

# start - Начальное значение последовательности
# stop - Конечное значение последовательности
# num - Количество элементов, default=50
# endpoint - включительно stop], default=True
# retstep - возвращает шаг между значениями и массив, default=False

print(x_range_float)
```
