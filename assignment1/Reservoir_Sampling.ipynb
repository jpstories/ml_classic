Резервуарный сэмплинг (Reservoir Sampling) — это семейство алгоритмов для выбора случайной выборки из потока данных, размер которого заранее неизвестен или слишком велик для сохранения в оперативную память.  

Он был создан для решения фундаментальной проблемы: как получить абсолютно честную, несмещенную случайную выборку из данных, если мы не знаем их общего объема и не можем загрузить их в память?

**Сферы применения:**

- **Обработка Big Data и логов:** Выборка 1000 случайных поисковых запросов из миллиардного потока Google в реальном времени.
- **Сетевой мониторинг:** Сбор статистики пакетов, проходящих через роутер, без остановки трафика.
- **Машинное обучение:** Обучение моделей на гигантских датасетах, которые считываются прямо из облачного хранилища (стриминг данных) без скачивания на диск.

Главные преимущества Reservoir Sampling:

- **Память \(O(k)\):** Ей абсолютно все равно, сколько элементов в датасете — 5 тысяч или 5 миллиардов. Потребление памяти константно и равно размеру выборки (\(k\)).
- **Время \(O(N)\):** Ровно один проход по данным. Алгоритм никогда не «зависнет» (в отличие от случайного поиска) и завершится гарантированно.

```python
import random

def reservoir_sampling_for_class(y_train, target_class, k=3):
    reservoir = []
    class_counter = 0  # Считаем только элементы нужного нам класса
    
    for idx, label in enumerate(y_train):
        if label == target_class:
            class_counter += 1
            
            # Элементы с 1 по k просто складываем в резервуар
            if class_counter <= k:
                reservoir.append(idx)
            else:
                # Для последующих элементов генерируем случайное число
                # Вероятность попасть в диапазон [0, k-1] равна k / class_counter
                j = random.randint(0, class_counter - 1)
                if j < k:
                    reservoir[j] = idx  # Заменяем старый элемент на новый
                    
    return reservoir

# Пример работы:
# y_train = [1, 2, 2, 3, 2, 4, 2, 5, 2] -> тут 5 двоек (индексы 1, 2, 4, 6, 8)
# result = reservoir_sampling_for_class(y_train, target_class=2, k=3)
# На выходе получим 3 случайных индекса двоек, например: [1, 6, 8]
```
