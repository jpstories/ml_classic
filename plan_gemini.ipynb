### Дополнения к "Требованиям к стандартам при обучении"

Твой наставник написал отличные 5 пунктов (Clean Code, SOLID, System Design, MLOps, Testing). Я добавляю еще два, которые в 2026 году отличают Senior Applied ML от просто хороших специалистов.

**6. Стандарт оптимизации Инференса (Model Deployment)** Обучить модель — это 20% работы Applied-инженера. 80% — заставить ее работать быстро и дешево.

- **Квантование (Quantization):** Ты должен понимать математику перевода весов из FP16/BF16 в INT8 или INT4 (AWQ, GPTQ) без критической потери точности.
    
- **Среды исполнения:** Забудь про запуск `.forward()` в ванильном PyTorch в продакшене. Изучай TensorRT (от NVIDIA), ONNX Runtime, vLLM (для трансформеров).
    
- **GPU-профилирование:** Научись пользоваться PyTorch Profiler и NVIDIA Nsight Systems. Ты должен видеть, где узкое место: в пересылке данных CPU->GPU (PCIe bottleneck) или в вычислительных ядрах (Compute bound).
    

**7. Программирование под GPU (Базовый уровень)** Элитарный Applied ML Engineer не боится спуститься на уровень железа.

- Изучи **Triton (от OpenAI)** или базовый **CUDA C++**. Умение написать кастомный kernel (например, свой оптимизированный слой FlashAttention или специфическую функцию активации), который работает в 5 раз быстрее стандартного PyTorch-кода — это то, за что платят сотни тысяч долларов в FAANG и стартапах.
    

### Стратегия перехода в Robotics AI (2027-2029)

Робототехника — это самая сложная сфера ИИ сегодня, потому что в ней ошибки в коде приводят к поломке железа стоимостью в миллионы. Как туда перейти:

1. **Связка Go/C++:** Твой бэкграунд в Go (MIT курс) будет супер-полезен для бэкенда. Но для Robotics (ROS2, управление контроллерами) тебе понадобится **C++17/20**. Робототехника не терпит Garbage Collectors. Начинай понемногу писать высокопроизводительный код на "плюсах".
    
2. **Симуляторы:** Начни изучать NVIDIA Isaac Sim. Современная робототехника учится не в реальном мире, а в фотореалистичных симуляторах с точной физикой, а затем переносится в реальность (Sim-to-Real transfer).
    
3. **Математика пространств:** Вспомни линейную алгебру (кватернионы, матрицы трансформации, кинематика). Без этого управлять манипулятором или дроном невозможно.
