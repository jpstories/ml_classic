## 1. Дорожная карта: Спринты (Месяцы 1–6)

Твоя цель на полгода — связать фундаментальную математику нейросетей с высоконагруженным бэкендом.

- **Месяцы 1–2: Низкоуровневый Deep Learning и CV**
    
    - **Материалы:** Stanford CS231n (Лабораторные 1–3) + плейлист Karpathy "Neural Networks: Zero to Hero".
        
    - **Фокус:** Аналитические градиенты, архитектура памяти (cache misses, broadcasting в NumPy).
        
    - **Итог:** Написанный с нуля на чистом Python/NumPy фреймворк автодифференцирования и CNN, обучающаяся на датасете CIFAR-10.
        
- **Месяцы 3–4: Трансформеры, LLM и Инференс**
    
    - **Материалы:** Stanford CS224N или создание своей LLM по лекциям Karpathy (nanoGPT).
        
    - **Фокус:** Multi-Head Self-Attention с нуля, RoPE (позиционные эмбеддинги), реализация KV-cache для ускорения генерации.
        
    - **Итог:** Собственная мини-LLM (уровня Llama-68M) с оптимизированным потреблением VRAM.
        
- **Месяцы 5–6: Распределенные системы и AI-Инфраструктура**
    
    - **Материалы:** MIT 6.5840 (Distributed Systems).
        
    - **Фокус:** Алгоритм консенсуса Raft, Sharded Key/Value Service на языке Go.
        
    - **Итог:** Отказоустойчивый кластер инференса, где балансировщик распределяет запросы к твоей LLM между несколькими нодами, и падение одной не прерывает работу.

## 3. Золотые стандарты инженерии (The 7 Pillars)

Твой код должен отличаться от скриптов дата-саентистов. Применяй эти 7 стандартов к каждому учебному проекту.

|**Стандарт**|**Суть для Applied ML**|**Инструменты и Практика**|
|---|---|---|
|**1. Clean Code**|Строгая типизация и документация каждой функции преобразования данных.|Google/Sphinx Docstrings, MyPy, Ruff, Black.|
|**2. Идемпотентность**|Пайплайны можно перезапускать 100 раз. Модули (загрузка, обучение) изолированы друг от друга (SOLID).|ООП, четкие интерфейсы компонентов.|
|**3. System Design**|Понимание потребления RAM/VRAM/CPU. Использование генераторов и векторизации.|NumPy, Pandas, отказ от циклов `for`.|
|**4. MLOps**|Воспроизводимость среды, данных и весов. Логирование каждого шага.|MLflow / W&B, DVC (Data Version Control).|
|**5. Тестирование**|Тесты пишутся для трех сущностей: кода логики, входных данных и поведения самой модели (edge cases).|PyTest, Great Expectations.|
|**6. Оптимизация**|Квантование весов, профилирование узких мест (CPU vs GPU bound), ускорение инференса.|TensorRT, vLLM, ONNX Runtime, PyTorch Profiler.|
|**7. GPU Code**|Умение писать кастомные низкоуровневые слои, минуя стандартные абстракции фреймворка.|Triton (OpenAI), базовый CUDA C++.|

## 4. Упаковка портфолио (GitHub Senior-уровня)

Твои учебные репозитории — это продукты с открытым исходным кодом.

- **Нейминг:** Никаких `course_work` или `test_task`. Только продуктовые названия (`raft-consensus-engine`, `nanogpt-kv-optimized`).
    
- **Культура коммитов:** Работа через ветки и Pull Requests. Использование Conventional Commits (`feat: ...`, `fix: ...`, `perf: ...`).
    
- **Структура репозитория:** Обязательное наличие `Makefile` (точка входа команд), `/tests`, конфигурационных файлов (Hydra/Pydantic).
    
- **Документация (README.md):** Включает архитектурные схемы (Mermaid.js), графики профилирования/бенчмарков и раздел Quickstart для запуска проекта в 3 команды.
    
- **CI/CD:** Настроенный пайплайн в GitHub Actions, который при каждом пуше прогоняет линтеры и тесты.
    

## 5. Инфраструктурная гигиена (Docker для GPU)

Контейнеризация — это не отдельный этап, а требование к каждому проекту с первого дня.

- **NVIDIA Container Toolkit:** Использование образов `cuda-base` и `cuda-runtime` для проброса видеокарты внутрь контейнера.
    
- **Multi-Stage Сборки:** Компиляция зависимостей в тяжелом builder-контейнере и перенос только готовых бинарников в легковесный финальный образ.
    
- **Управление Volumes:** Вынос тяжелых весов моделей (10+ ГБ) и датасетов за пределы Docker-образа (через Bind Mounts или S3).
    
- **Оркестрация:** Запуск связки "Модель + Векторная БД + Балансировщик" через `docker-compose up`.
    

> **Главный совет на старте:** Будет тяжело, особенно когда начнется отладка математики в тензорах. Не пытайся учить всё сразу. Сначала заставь математику работать, потом сделай код красивым, затем оберни в Docker, и в самом конце — напиши CI/CD. Двигайся итеративно.
