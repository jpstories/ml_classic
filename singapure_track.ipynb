## Визовая стабильность и зависимые (Dependant's Pass)

|Параметр|Детали|Стабильность для семьи|
|---|---|---|
|**Employment Pass (EP)**|Минимальная зарплата: **S$6,000/мес** (с 2027, растёт с возрастом до S$11,500 для mid-40s). Валидность: **2 года** (первоначально), до **3 лет** при продлении.|✅ **Высокая**: EP — основной трек для квалифицированных специалистов. COMPASS scoring (40+ баллов) даёт предсказуемость. asiaone+2|
|**Dependant's Pass (DP)**|Супруга и дети до 21 года получают DP. **Супруга может работать** через Letter of Consent (LOC) или свой EP. Дети — доступ к международным школам.|✅ **Высокая**: DP выдаётся автоматически при EP > S$6,000. Супруга может работать (LOC или EP). kurums+1|
|**ONE Pass (AI & Tech Track, с 2027)**|Зарплата: **S$30,000/мес** (включая equity). Валидность: **5+5 лет**. **Супруга может работать без отдельного пропуска**.|✅✅ **Очень высокая**: Для senior-инженеров (AI/ML, CV, robotics) — **лучший трек**. Долгосрочная стабильность, гибкость (можно работать в нескольких компаниях). pinsentmasons+2|
|**Tech.Pass (до конца 2026)**|Зарплата: S$20,000/мес. Валидность: 2+2 года. **Заменяется на ONE Pass с 2027**.|⚠️ **Средняя**: Если подаётесь до конца 2026 — можно получить, но потом переход на ONE Pass. visasupdate+1|
|**Permanent Residence (PR)**|Подача после **2-3 лет на EP**. Одобрение — **дискреционное** (зависит от зарплаты, вклада, семьи).|⚠️ **Средняя**: Нет гарантии, но при высокой зарплате (S$10k+) и семье шансы **выше**. [kurums](https://kurums.com/singapore-work-visa-guide-expats/)|
