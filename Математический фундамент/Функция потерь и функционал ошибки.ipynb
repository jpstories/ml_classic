# Как измеряется промах

Это судья, который измеряет, насколько модель ошиблась в предсказании, при условии что есть правильные ответы для сравнения.
Нужны для того, чтобы модель понимала, **хорошо она работает или плохо**, и как ей исправиться.

---

## 1. Функция потерь — $L(y, z)$ (Loss Function)
Определяет величину ошибки алгоритма на **одном конкретном объекте**.

* **Что принимает на вход**: 
  * $y$ — истинный ответ (таргет) для одного объекта.
  * $z$ — прогноз модели для этого же объекта (часто записывается как $a(x)$).
* **Масштаб**: Локальный (вычисляется для одной строки датасета).
* **Вопрос, на который отвечает**: *«Насколько сильно модель ошиблась на конкретном треке/пациенте?»*

### 📝 Пример (Квадрат ошибки в регрессии)
Для конкретного $i$-го объекта функция потерь выглядит так:
$$L(y_i, a(x_i)) = (a(x_i) - y_i)^2$$

---

## 2. Функционал ошибки — $Q(a, X)$ (Cost / Risk Function)
Определяет общее качество работы модели на **всей выборке данных**.
Чаще всего — это просто **среднее арифметическое** от всех индивидуальных функций потерь.
То есть мы складываем все индивидуальные промахи и делим на их количество.

Цель обучения с учителем — крутить внутренние настройки модели до тех пор, пока общее число $Q$ не станет как можно ближе к нулю.

* **Что принимает на вход**:
  * $a$ — сам алгоритм (модель с её текущими весами).
  * $X$ — вся обучающая выборка (все объекты и правильные ответы).
* **Масштаб**: Глобальный (вычисляется по всему датасету).
* **Математическая суть**: Это **функционал**, потому что его аргументом является целая функция (алгоритм $a$), а на выходе получается число. Обычно это просто среднее арифметическое функций потерь по всем $l$ объектам:
$$Q(a, X) = \frac{1}{l} \sum_{i=1}^{l} L(y_i, a(x_i))$$
* **Вопрос, на который отвечает**: *«Насколько хорошо модель работает по всей базе данных в среднем?»* Именно функционал ошибки минимизирует алгоритм градиентного спуска во время обучения.

### 📝 Пример (Метрика MSE)
Если мы подставим квадрат ошибки из пункта 1 в формулу среднего значения, функционал ошибки превратится в знаменитый **MSE** (Mean Squared Error):
$$Q(a, X) = \frac{1}{l} \sum_{i=1}^{l} (a(x_i) - y_i)^2$$

---

## ⚙️ Важное отличие: Регуляризация (Штраф за сложность)
Различие становится окончательно понятным при добавлении регуляризации ($R(w)$) для защиты от переобучения. Штраф за слишком большие веса модели добавляется **только в функционал ошибки $Q$**, но никогда не добавляется в локальную функцию потерь $L$.

**Полный функционал ошибки (например, Ridge-регрессия):**
$$Q(a, X) = \underbrace{\frac{1}{l} \sum_{i=1}^{l} L(y_i, a(x_i))}_{\text{Средняя потеря на выборке}} + \underbrace{\lambda \sum_{j=1}^{d} w_j^2}_{\text{Регуляризатор (Штраф)}}$$

---

## 📌 Шпаргалка для быстрого запоминания
* **$L(y, z)$** — Ошибка на **одном** примере (строке).
* **$Q(a, X)$** — Средняя ошибка на **всех** примерах + штрафы.
