Смещение (Bias)

- **Что это:** Ошибка из-за того, что модель слишком простая и делает грубые предсказания.

- **Простыми словами:** Модель «не доучила» правила и упускает важные связи в данных.

- **Результат:** Недообучение (underfitting), когда модель плохо работает и на обучающих, и на новых данных.

Разброс (Variance)

- **Что это:** Ошибка из-за того, что модель слишком сложная и реагирует на малейший шум в данных.

- **Простыми словами:** Модель слишком сильно подстраивается под конкретный пример обучения.

- **Результат:** Переобучение (overfitting), когда на старых данных всё идеально, а на новых — большие ошибки.

Баланс (Trade-off)

- Уменьшая смещение (делая модель сложнее), мы увеличиваем разброс.

- Уменьшая разброс (упрощая модель), мы увеличиваем смещение.

- Задача инженера — найти золотую середину, чтобы общая ошибка была минимальной.

#### Как бороться с высоким Bias (Недообучением)

Высокий bias означает, что модель слишком простая. Ее нужно **усложнить** и дать ей больше свободы.

- **Увеличить сложность модели**: Добавить слои в нейросеть или увеличить глубину деревьев решений.
- **Добавить новые признаки (Features)**: Создать новые параметры из существующих или собрать дополнительные данные.
- **Использовать полиномиальные признаки**: Добавить квадраты или кубы признаков, чтобы модель могла находить нелинейные зависимости.
- **Уменьшить регуляризацию**: Ослабить штрафы (например, уменьшить коэффициенты L1/L2-регуляризации), которые сдерживают модель.

#### Как бороться с высоким Variance (Переобучением)

Высокий variance означает, что модель зазубрила обучающую выборку. Ее нужно **упростить** или дать больше данных для обобщения.

- **Добавить больше обучающих данных**: Больший объем данных мешает модели подстраиваться под случайный шум.
- **Применить регуляризацию**: Использовать **L1 (Lasso)** или **L2 (Ridge)** штрафы, чтобы уменьшить веса признаков и упростить модель.
- **Уменьшить количество признаков**: Удалить неважные или сильно коррелирующие параметры (Feature Selection).
- **Использовать ансамбли (Bagging)**: Объединить несколько сложных моделей. Например, **Random Forest** эффективно снижает variance.
- **Ранняя остановка (Early Stopping)**: Прекратить обучение нейросети, как только ошибка на валидационных данных начала расти.
