**Gradient Descent** — это алгоритм обучения модели. Он берет **Функционал ошибки** и крутит веса модели так, чтобы свести его к минимуму. Этот подход используется во всех задачах обучения с учителем: от регрессии до мультикатегориальной классификации.

---

## 🔄 Пошаговый алгоритм: как учится модель
Внутри библиотек и фреймворков (будь то CatBoost, Scikit-learn или PyTorch) при запуске обучения всегда крутится следующий цикл:

1. **Инициализация**: Модель выставляет свои внутренние веса совершенно случайно. Ошибка гигантская — мы стоим на вершине горы.
2. **Прямой ход (Предсказание)**: Модель делает предсказания по выборке, а функционал ошибки $Q$ считает общую величину промаха.
3. **Расчет градиента**: Математический движок берет производную от функционала ошибки. Производная отвечает на вопрос: *«Если мы чуть-чуть изменим веса модели, ошибка поползет вверх или вниз?»*. Полученный градиент указывает крутое направление вверх (к вершине ошибки).
4. **Шаг (Обновление весов)**: Модель меняет свои веса в противоположную от градиента сторону. Она делает «шаг вниз по склону горы».
5. **Повторение**: Шаги 2–4 повторяются тысячи раз (эти шаги называют *итерациями* или *эпохами* обучения), пока модель не спустится в долину, где ошибка перестает уменьшаться.

---

## ⚡ Важнейший параметр: Скорость обучения (Learning Rate)
Это **размер шага**, который модель делает за один раз на склоне горы. Его настраивает дата-сайентист:

* **Слишком большой шаг (высокий Learning Rate):** Робот гигантскими прыжками скачет по горе, может легко перепрыгнуть долину с минимальной ошибкой и улететь на соседний склон. Обучение расходится.
* **Слишком маленький шаг (низкий Learning Rate):** Модель спускается микроскопическими шажками. Обучение займет вечность, к тому же модель может безнадежно застрять в первой же мелкой ямке (локальный минимум), приняв её за финальное дно долины.
