Вся линейная алгебра строится на трех объектах: 
- скалярах
- векторах
- матрицах.

---

Скаляр — это просто **одно число**. В машинном обучении скалярами обычно бывают коэффициенты, скорость обучения (learning rate) или финальная ошибка модели.

- **Математическая запись:** x e {R} (число x принадлежит множеству вещественных чисел).
- **В Python:** Это обычный тип `int` или `float`.

```python
# Скаляр в Python
learning_rate = 0.01
```

---

Вектор — это **упорядоченный список чисел**. В ML вектор — это главный способ описать один объект.

Представьте, что мы предсказываем цену квартиры. Один объект (квартира) описывается вектором признаков: $\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} \in \mathbb{R}^3$   
Где x1 — площадь, x2 — число комнат, x3 — этаж. По умолчанию в математике векторы всегда записываются как **столбцы** размера (n x 1), где n — количество элементов (размерность).

- **В Python:** Создается через одномерный массив `np.array`.

```python
import numpy as np

# Вектор признаков квартиры (размерность 3)
x = np.array([54, 2, 7]) 
print(x.shape)  # Выведет (3,), в NumPy одномерный массив универсален
```

---

Матрица — это **двумерная таблица чисел**, состоящая из строк и столбцов. В ML матрица — это весь ваш датасет (набор объектов).

Если у нас есть данные о 3-х разных квартирах, мы объединяем их векторы в одну матрицу \(X\):  
$$X = \begin{bmatrix} 54 & 2 & 7 \\ 38 & 1 & 12 \\ 90 & 3 & 3 \end{bmatrix}$$

- В Python: Создается как двумерный массив.

```python
# Матрица данных (3 объекта, у каждого по 3 признака)
X = np.array([[54, 2, 7], [38, 1, 12], [90, 3, 3]])
print(X.shape)  # Выведет (3, 3) -> (строки, столбцы)
```

Базовые операции: Сложение и Умножение на число

С этими объектами можно делать простые арифметические действия. Главное правило — они делаются **поэлементно**.

1. **Умножение вектора/матрицы на скаляр:** каждое число внутри умножается на этот скаляр.
    - Если вектор цен умножить на \(1.1\), мы увеличим все цены на 10%.
2. **Сложение:** складывать можно только векторы одинаковой длины или матрицы одинакового размера. Элементы, стоящие на одинаковых позициях, просто складываются между собой.

```python
# Пример поэлементного сложения векторов в NumPy
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print(v1 + v2)  # Выведет: [5, 7, 9]
print(v1 * 2)   # Выведет: [2, 4, 6] (умножение на скаляр)
```

Реализация в Python (NumPy)

В `NumPy` для скалярного произведения есть специальный оператор `@` или функция `np.dot()`

```python
import numpy as np

# Задаем векторы
x = np.array([50, 10])
w = np.array([2, -3])

# Способ 1: через оператор @ (наиболее современный и частый)
result_1 = x @ w

# Способ 2: через функцию np.dot
result_2 = np.dot(x, w)

print(f"Результат через @: {result_1}")  # Выведет: 70
print(f"Результат через np.dot: {result_2}")  # Выведет: 70
```

В NumPy транспонировать массив можно очень просто — вызвав атрибут `.T` у переменной матрицы.

```python
import numpy as np

# Создаем матрицу 2x3
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Транспонируем ее
A_transposed = A.T

print("Исходная матрица A:")
print(A)
print(f"Размерность: {A.shape}\n")

print("Транспонированная матрица A^T:")
print(A_transposed)
print(f"Размерность: {A_transposed.shape}")
```

Зачем это нужно в ML?

В реальных задачах признаки объектов часто хранятся в строках, но формулам для вычислений (например, в нейросетях) данные нужны в виде столбцов. Транспонирование — это главный инструмент для «подгонки» размерностей матриц друг под друга, чтобы их в принципе можно было умножить.

Реализация в Python (NumPy)

В коде за это отвечает уже знакомый нам оператор `@` (или функция `np.dot()` / `np.matmul()`).

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [5, 2],
    [2, 3],
    [1, 4]
])

# Матричное умножение
C = A @ B

print("Результат умножения матриц C:")
print(C)
print(f"Размерность: {C.shape}") # Выведет (2, 2)
```

```python
import numpy as np
I = np.eye(3) # Создает единичную матрицу 3х3
print(I)

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

C = A @ I
print(C)

```

```python
A = np.array([
    [1, 2], 
    [3, 4]
])
A_inv = np.linalg.inv(A) # Нахождение обратной матрицы

print("Обратная матрица A^-1:")
print(A_inv)

print("\nПроверка A @ A^-1 (должна получиться единичная):")
print(np.round(A @ A_inv)) # Округляем, чтобы убрать микро-погрешности float
```

Реализация в Python (NumPy)

В NumPy есть два способа решить такую систему. Второй способ — np.linalg.solve — математически более стабилен и работает быстрее, чем явное вычисление обратной матрицы.

```python
import numpy as np

A = np.array([
    [1, 2], 
    [3, 4]
])

b = np.array([5, 11])

# Способ 1: Прямо по формуле x = A^-1 @ b
x_method1 = np.linalg.inv(A) @ b

# Способ 2: Оптимизированный встроенный решатель (Рекомендуемый!)
x_method2 = np.linalg.solve(A, b)

print(f"Решение через обратную матрицу: {x_method1}") # Выведет [1. 2.]
print(f"Решение через np.linalg.solve:  {x_method2}") # Выведет [1. 2.]
# Проверка: 1*1 + 2*2 = 5; 3*1 + 4*2 = 11. Всё верно!
```

#### Цикл градиентного спуска на NumPy (.fit)

```python
import numpy as np

# 1. Готовим данные: x1 = 50 кв.м, x2 = 2 км от центра
X = np.array([
    [50, 2],
    [35, 12],
    [90, 5],
    [70, 15]
])
y = np.array([12.1, 5.3, 19.5, 11.2]) # цена

# Добавляем столбец из единиц для свободного коэффициента (w_0)
ones = np.ones((X.shape[0], 1))
#[[1.], [1.], [1.], [1.]]

X_with_bias = np.hstack((ones, X))
# [ 
# [ 1. 50.  2.], 
# [ 1. 35. 12.], 
# [ 1. 90.  5.], 
# [ 1. 70. 15.]
# ]
# shape: 4 x 3


m = X_with_bias.shape[0]  # Количество строк (4)

# 2. Инициализируем веса СЛУЧАЙНЫМИ числами или нулями
# У нас 3 веса: w_0 (bias), w_1 (площадь), w_2 (расстояние)
w = np.zeros(X_with_bias.shape[1])
# [0. 0. 0.] - ВЕСА

# 3. Настройки градиентного спуска
learning_rate = 0.0001  # Маленький шаг, чтобы обучение шло стабильно
epochs = 5000           # Сколько раз мы пройдем по всему датасету

print("Начальные веса (модель еще ничего не знает):", w)
print("-" * 50)

# 4. Главный цикл обучения
for epoch in range(1, epochs + 1):
    # Шаг А: Делаем предсказание текущими весами (Матричное умножение!)
    predictions = X_with_bias @ w
    print('predictions: ', predictions)
    print('y: ', y)
    # Шаг Б: Считаем текущую ошибку MSE (для визуализации)
    error = predictions - y
    print('error: ', error)
    loss = np.mean(error ** 2)
    print('loss: ', loss)
    
    # Шаг В: Вычисляем градиент по матричной формуле (Матанализ в действии)
    # Слэш / m нужен, чтобы отнормировать градиент по количеству строк
    gradient = (2 / m) * (X_with_bias.T @ error)

    print('gradient: ', gradient)
    
    # Шаг Г: Обновляем веса (шаг в сторону антиградиента)
    w = w - learning_rate * gradient
    print('new w: ', w)
    
    # Печатаем прогресс каждые 500 эпох
    if epoch % 500 == 0 or epoch == 1:
        print(f"Эпоха {epoch:4d} | Ошибка MSE: {loss:8.4f} | Веса: {w}")

print("-" * 50)
print("Итоговые веса после градиентного спуска:", w)
```
