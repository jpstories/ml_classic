Основа для статистики и маш. обучения, оценка того что предсказание модели правильное

```python
import numpy as np

# Наш признак "Площадь"
area = np.array([50, 35, 90, 70])

# Стандартизация на чистом NumPy
mu = np.mean(area)
sigma = np.std(area)

area_scaled = (area - mu) / sigma
print("Стандартизированная площадь:", area_scaled)
# Теперь данные сбалансированы вокруг нуля!
```
