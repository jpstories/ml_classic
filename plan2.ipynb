## План обучения: От кода к нейронкам (6–8 месяцев)

Предполагается, что ты будешь тратить 15–20 часов в неделю.

### 1. Математический минимум (Месяц 1)

Не пытайся вызубрить университетский курс мехамата — перегоришь. Тебе нужна математика ровно до того уровня, чтобы понимать, как работает функция потерь (Loss) и оптимизатор.

- **Что учить:**
    
    - Линейная алгебра (векторы, матрицы, скалярное произведение, матричное умножение).
        
    - Матанализ (производная, градиент, цепное правило — Chain Rule).
        
    - Основы теории вероятностей и статистики (распределения, матожидание, дисперсия, p-value).
        
- **Домашка:** Написать линейную и логистическую регрессию чисто на Python с `NumPy` (без готовых ML-библиотек), реализовав градиентный спуск вручную.
    

### 2. Классическое ML и работа с данными (Месяц 2)

До нейронок нужно понять, как работать с табличными данными — это всё ещё 50% задач в финтехе и e-commerce.

- **Что учить:**
    
    - Инструменты: `Pandas`, `Scikit-Learn`.
        
    - Алгоритмы: Решающие деревья, Случайный лес (Random Forest), Градиентный бустинг (`CatBoost` от Яндекса, `XGBoost`).
        
    - Концепции: Кросс-валидация, переобучение (overfitting), метрики (Precision, Recall, F1, ROC-AUC для классификации; RMSE, MAE для регрессии).
        
- **Домашка:** Зайти на Kaggle, взять популярное табличное соревнование (например, кредитный скоринг или предсказание цен), сделать качественную очистку данных (Feature Engineering) и обучить `CatBoost`. Заслать результат в Leaderboard.
    

### 3. Deep Learning и PyTorch (Месяцы 3–4)

Забудь про TensorFlow, индустрия окончательно перешла на PyTorch. Здесь начинается настоящая магия.

- **Что учить:**
    
    - Основы PyTorch: тензоры, `autograd`, `nn.Module`, `DataLoader`.
        
    - Архитектуры: Многослойный перцептрон (MLP), Сверточные сети (CNN — ResNet, ConvNeXt) для CV, рекуррентные (LSTM) — факультативно.
        
    - Оптимизаторы: Adam, AdamW.
        
- **Домашка:** Обучить CNN для классификации изображений с нуля. Настроить трекинг экспериментов (чтобы графики Loss и Accuracy рисовались красиво) с помощью `Weights & Biases (W&B)`.
    

### 4. LLM, RAG и GenAI — Твоя золотая жила (Месяцы 5–6)

Это то, за что сейчас платят больше всего. Компании хотят внедрять ИИ в свои продукты, и им нужны инженеры, умеющие работать с большими языковыми моделями.

- **Что учить:**
    
    - Архитектура Transformer (в деталях понять механизм Attention).
        
    - Экосистема Hugging Face (`transformers`, `peft`, `datasets`).
        
    - RAG (Retrieval-Augmented Generation): как скармливать модели корпоративные документы.
        
    - Векторные базы данных (`Qdrant`, `Milvus`, `pgvector`).
        
    - Fine-tuning: LoRA, QLoRA (как дообучить LLM на одной видеокарте).
        
- **Домашка:** Собрать end-to-end RAG-систему на своих документах (например, бот, отвечающий по логам твоего старого проекта). **Важно:** напиши пайплайн без LangChain, на чистом Python + API модели + Векторная БД. LangChain хорош для прототипов, но в проде он превращается в неподдерживаемое месиво.
    

### 5. MLOps и Production (Месяц 7)

Твоя родная территория, где ты порвёшь чистых дата-саентистов. Задача — научиться правильно деплоить модели.

- **Что учить:**
    
    - `MLflow` (реестр моделей и версионирование).
        
    - Инференс: `vLLM` (для языковых моделей), `Triton Inference Server` от NVIDIA (для CV и классических моделей).
        
    - Оптимизация: Квантизация моделей (AWQ, GGUF), конвертация в `ONNX` / `TensorRT`.
        
- **Домашка:** Взять обученную на шаге 4 модель, квантовать её до 4-bit, обернуть в Docker-контейнер с `FastAPI` (или `Triton`) и написать нагрузочный тест, замерив пропускную способность (RPS) и задержку (Latency).
    


## 🛠 Железо, Инструменты и Лайфхаки

1. **Где брать GPU (видеокарты):** Твоего макбука не хватит.
    
    - Для старта: Google Colab или Kaggle Notebooks (там дают бесплатные GPU, но с ограничениями).
        
    - Для реальных проектов: сервисы почасовой аренды серверов **Vast.ai** или **RunPod**. За $0.30–$0.50 в час ты получаешь мощнейшую RTX 3090/4090, чего хватит для 90% учебных задач.
        
2. **Помощники:** Используй Cursor или GitHub Copilot. При переходе в ML тебе придется писать много boilerplate-кода для работы с размерностями тензоров. Нейронки отлично пишут нейронки.
    
3. **Не читай статьи на Medium, читай документацию и пейперы.** В ML всё устаревает за 3 месяца. Документация PyTorch, Hugging Face и статьи на arXiv — твои лучшие друзья.
