### 1. Соблюдайте четкую структуру ячеек

Блокнот Jupyter может быстро превратиться в хаос. Приучите себя разделять код на логические блоки
(ячейки). Хорошим стандартом считается следующая последовательность ячеек:

1. **Ячейка импортов:** Все `import` всегда находятся в самой первой ячейке. Никогда не импортируйте
   библиотеки в середине блокнота.
2. **Глобальные константы:** Гиперпараметры (batch size, learning rate, количество эпох) выносятся
   наверх.
3. **Загрузка и подготовка данных (ETL):** Ячейки с функциями для чтения датасета и создания
   `DataLoader`.
4. **Определение архитектуры (Модель):** Класс вашей нейросети.
5. **Функции обучения и валидации:** Логика цикла обучения (`train_step`, `val_step`).
6. **Главный цикл обучения (Main Loop):** Ячейка, где вызывается процесс тренировки и логируется
   Loss.
7. **Оценка результатов (Инференс):** Построение графиков, визуализация предсказаний на тестовых
   картинках.


### 2. Контролируйте случайность (Random Seed)

Нейросети инициализируются случайными весами, а данные перемешиваются. Чтобы вы могли воспроизвести
свои результаты (или понять, почему при изменении одной строчки кода точность упала), вам нужно
зафиксировать генераторы случайных чисел.

В начале каждого блокнота выполняйте подобный код (пример для PyTorch):


```python
import torch
import numpy as np
import random


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    # Если используете GPU:
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)


set_seed(42)
```
