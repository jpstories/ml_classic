Чек-лист для адаптации резюме под Японию

Японские англоязычные ИИ-рекрутеры смотрят на резюме специфически. Чтобы пройти первичный скрининг: [[1](https://www.jac-international.jp/job-seekers/how-to-write-a-cv-in-english-for-japan-ai-smart-guide-for-bilingual-talent/using-ai-to-improve-your-cv)]

- **Уберите информацию о визе из шапки:** Не нужно писать «Нужна виза». Просто укажите в поле _Location_: `Current: Moscow, Russia | Open to relocation to Tokyo, Japan`. Местные компании по умолчанию спонсируют рабочую визу для ИИ-кадров. [[1](https://japan-dev.com/ml-jobs-in-japan), [2](https://global.bloomtechcareer.com/media/contents/ai-engineer-career-in-japan-salary-visa-success/)]
- **Покажите Production-опыт:** Японский рынок страдает от обилия ученых-теоретиков без практики. Сделайте акцент не просто на обучении моделей, а на их деплое: `TensorRT`, `ONNX`, `Triton Inference Server`, `C++ (для эмбеддед-систем/камер)`.
- **Добавьте метрики бизнеса:** Вместо «Занимался CV» пишите по системе STAR: _«Оптимизировал модель детекции объектов (YOLOv8) под Edge-устройства, что снизило задержку (latency) на 40% и сэкономило компании $20k в месяц на серверной инфраструктуре»_.
- **Подсветите Generative AI в CV:** Если вы работали с `Stable Diffusion`, `ControlNet`, `multimodal LLM (image2text / text2image / text2video)`, вынесите это на самый верх CV. В 2026 году это самый растущий сектор по зарплатам. [[1](https://www.daijob.com/en/jobs/search_result?jt[]=614&job_search_form_hidden=1&job_types[]=614), [2](https://gaijinhunterjp.com/guides/data-ai)]
- **Прикрепите GitHub/HuggingFace:** Ссылки на код или ваши собственные инференс-демо (например, на Gradio/Streamlit) ускоряют получение оффера в два раза
