### Главная цель задачи

Разбить исходное множество объектов на группы (кластеры) так, чтобы объекты внутри одного кластера   
были максимально похожи, а объекты из разных групп — максимально отличались друг от друга.

Популярные алгоритмы кластеризации

- **K-Means (K-средних)**: Делит данные на заранее заданное число кластеров (K). Алгоритм выбирает   
случайные центры (центроиды), привязывает к ним ближайшие точки, а затем пересчитывает центр группы,   
повторяя процесс до сходимости.

- **DBSCAN**: Кластеризует по плотности точек. Он отлично находит группы сложной геометрической формы  
 и автоматически определяет шум (выбросы), не требуя заранее указывать количество кластеров.

- **Иерархическая кластеризация**: Строит дерево сходства (дендрограмму). Она последовательно либо   
объединяет мелкие группы в одну большую (агломеративная), либо разбивает один большой кластер на   
мелкие (дивизивная).

### Метрики (как оценивают качество)

Поскольку «правильных ответов» нет, используются два типа метрик:

- **Внутренние метрики**: Оценивают структуру кластеров без сторонней информации. Например,   
_Силуэт (Silhouette Score)_ проверяет, насколько объект близок к своему кластеру по сравнению с чужими  
 (идеал — ближе к 1).

- **Внешние метрики**: Используются только на тестовых или синтетических данных, где реальные метки всё  
 же известны. Например, _Adjusted Rand Index (ARI)_ измеряет сходство полученного разбиения с эталонным.

### Основные сферы применения

- **Сегментация клиентов**: Разделение покупателей на группы по поведению для таргетинга рекламы.
- **Сжатие изображений**: Уменьшение количества цветов на фото за счет группировки похожих оттенков.
- **Поиск аномалий**: Выявление мошеннических транзакций в банках, которые сильно выбиваются из стандартных кластеров.
