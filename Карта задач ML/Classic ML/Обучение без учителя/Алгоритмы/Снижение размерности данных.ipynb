**PCA (Principal Component Analysis / Метод главных компонент)** — это один из самых популярных алгоритмов обучения без учителя для **снижения размерности данных**.

Представьте, что у вас в таблице не 3 признака (как в RFM), а 150 (возраст, модель телефона, время в приложении, клики, категории товаров и т.д.).  
Построить 3D-график уже не получится, а алгоритмы кластеризации вроде K-Means начинают путаться и работать медленно.   
Это называют **«проклятием размерности»**.

PCA берет эти 150 признаков и «сжимает» их, например, в 2 или 3 совершенно новых признака (их называют _главными компонентами_).   
При этом алгоритм старается **сохранить максимум полезной информации** (дисперсии) из исходного датасета.

---

Жизненный пример для понимания

Представьте **3D-объект** — например, заварочный чайник.   

Если вы сфотографируете его сбоку, вы получите **2D-картинку** (снизили размерность с 3 до 2),   
но по ней всё равно легко понять, что это чайник.   

Вы видите и ручку, и носик. Фотография сбоку — это и есть _первая главная компонента_,   
которая сохранила максимум  информации об объекте. А вот если сфотографировать чайник строго сверху,  
 получится просто кружок — эта проекция потеряла почти всю информацию. PCA ищет именно такие ракурсы  
  (компоненты), которые сохраняют «форму» данных лучше всего.
