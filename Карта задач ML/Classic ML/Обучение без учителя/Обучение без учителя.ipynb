I. Обучение без учителя (Supervised Learning)
Правильных ответов в данных нет.  
Модель самостоятельно ищет скрытую структуру, логику и закономерности.

ML-инженер использует эти алгоритмы, когда нужно навести порядок в хаосе данных:   
- сгруппировать миллионы клиентов
- найти аномальные транзакции
- сжать огромную таблицу признаков до понятного размера.

В коммерческом продакшене обучение без учителя делится на 3 главные задачи

- 👥 **Кластеризация (Clustering)**
    - _Суть:_ Группировка похожих объектов в группы (кластеры), которые заранее не определены человеком.
    - _Пример:_ Разделение слушателей на группы по вкусам (например, «любители подкастов», «фанаты ночного техно»).
- 🗜️ **Уменьшение размерности (Dimensionality Reduction)**
    - _Суть:_ Сжатие огромного количества признаков объекта до компактного вектора без потери важного смысла.
    - _Пример:_ Сжатие аудио-характеристик трека в вектор из 128 чисел для мгновенного поиска похожих аудиозаписей.
- 🚨 **Поиск аномалий (Anomaly Detection)**
    - _Суть:_ Выявление редких объектов, которые критически отличаются от нормального поведения большинства.
    - _Пример:_ Обнаружение ботов, которые имитируют прослушивания и накручивают счетчики стриминга.
- 🔗 **Поиск ассоциативных правил (Association Rule Learning)**
    - _Суть:_ Поиск скрытых правил в поведении пользователей («кто берет товар А, также берет товар Б»).
    - _Пример:_ Если пользователь добавляет в плейлист этот трек, он с вероятностью 85% добавит и следующий.

Карта алгоритмов: что и когда выбирать

|Алгоритм|Главная фишка|Когда идеален|Когда бесполезен|
|---|---|---|---|
|**K-Means**|Насильно режет данные на K округлых групп|Бизнес требует четкого деления на K сегментов; данные распределены плотным облаком|Кластеры имеют сложную форму (линии, полумесяцы, кольца)|
|**DBSCAN**|Ищет группы по плотности точек, находит аномалии|Вы не знаете число кластеров; группы сложной формы; в данных много шума|Данные распределены равномерно; у кластеров сильно разная плотность|
|**PCA**|Сжимает данные по прямым линиям (линейно), работает мгновенно|Нужно ускорить обучение моделей; убрать лишние признаки; сохранить максимум информации|Данные связаны сложными нелинейными зависимостями (спирали, сферы)|
|**t-SNE**|Разворачивает сложные структуры ради красивой 2D/3D картинки|Нужно сделать презентацию для бизнеса и красиво показать «островки» данных|Нельзя использовать как признаки для обучения других моделей; очень медленный|

🚀 Пошаговый пайплайн решения задачи (Ваш чек-лист)

Если перед вами лежит массив сырых данных без ответов, двигайтесь по этой цепочке:

1. **Предобработка**: Очистите данные от явных ошибок.  

2. **Масштабирование (Критично!)**: Всегда используйте `StandardScaler` или `MinMaxScaler`. Без этого кластеризация пойдет по признаку с самыми большими цифрами.  

3. **Снижение размерности (если признаков > 15–20)**: Примените **PCA**, чтобы сократить количество колонок, сохранив 85–90% информации.  

4. **Кластеризация**:  
    - Запустите **K-Means**, подобрав число групп методом «локтя».
    - Если подозреваете наличие шума или сложные формы — попробуйте **DBSCAN**.  

5. **Валидация**: Оцените структуру математически с помощью метрики **Силуэта (Silhouette Score)**.  

6. **Визуализация**: Сожмите результаты в 2D с помощью **t-SNE** и раскрасьте точки полученными кластерами для наглядного отчета.  

7. **Интерпретация**: Посчитайте средние значения признаков для каждого кластера. Опишите понятным языком, что это за группы (например: _"Кластер 1 — это VIP-клиенты, покупают часто и дорого"_).

Подготовка данных: Где больше рутины?

- **Обучение с учителем (Сложнее):** Огромная боль — это сбор и разметка данных.   
Чтобы обучить модель распознавать котиков, человеку нужно вручную разметить миллион  
 картинок. Это долго, дорого и скучно.  

- **Обучение без учителя (Проще):** Данные можно брать «как есть» (сырые логи,   
транзакции, тексты). Вам не нужно нанимать людей для разметки, алгоритм сам начнет  
 искать в них закономерности.

Настройка и стабильность алгоритмов

- **Обучение с учителем (Проще):** Модели более стабильны. Если вы обучили нейросеть  
 или случайный лес (Random Forest), они будут выдавать предсказуемый результат.  

- **Обучение без учителя (Сложнее):** Алгоритмы невероятно капризны к случайности.   
Вы только что видели на практике: изменили параметр `eps` в DBSCAN с 0.15 на 0.08,   
и модель из одного гигантского кластера выдала 1000 точек шума. Чуть поменялся   
масштаб данных — и вся кластеризация развалилась.

Интерпретация: Как объяснить результат человеку?

- **Обучение с учителем (Проще):** «Модель предсказала, что этот заемщик — мошенник   
с вероятностью 88%». Всем всё понятно.  

- **Обучение без учителя (Сложнее):** Алгоритм выдал вам «Кластер №4». Вы открываете  
 его, а там смесь из разных признаков. Вам нужно проявить детективные способности и  
  бизнес-интуицию, чтобы понять и доказать коллегам, почему эти объекты оказались вместе.  

Итог:  

- Обучение с учителем сложнее на старте — нужно потратить много сил и денег на создание   
качественной обучающей выборки (датасета с ответами).  

- Обучение без учителя сложнее в процессе и в конце — запрограммировать его легко, но  
 превратить хаотичные группы точек в реальную пользу для бизнеса — это высший пилотаж и   
 творчество.
