# ФУНКЦИИ ОШИБКИ

```python
ФУНКЦИИ ОШИБКИ (LOSS FUNCTIONS)
│
├── ЗАДАЧИ КЛАССИФИКАЦИИ (Оптимизируют предсказание вероятностей)
│   │
│   ├── Бинарная классификация (2 класса)
│   │   └── Binary Cross-Entropy / Log Loss (Штрафует экспоненциально за уверенную ошибку)
│   │
│   ├── Многоклассовая классификация (3 и более классов)
│   │   ├── Categorical Cross-Entropy (Для target-векторов в виде One-Hot кодирования)
│   │   └── Sparse Categorical Cross-Entropy (Для target-векторов в виде простых индексов 0, 1, 2...)
│   │
│   ├── Для данных с сильным дисбалансом классов
│   │   └── Focal Loss (Заставляет модель фокусироваться на редких и сложных объектах)
│   │
│   └── Для методов опорных векторов (SVM)
│       └── Hinge Loss (Штрафует даже за верные ответы, если они близко к границе классов)
│
├── ЗАДАЧИ РЕГРЕССИИ (Оптимизируют предсказание непрерывных чисел)
│   │
│   ├── На основе Среднего Арифметического ошибок
│   │   ├── MSE (Mean Squared Error) ──► Квадратичный штраф. Очень чувствительна к выбросам
│   │   └── MAE (Mean Absolute Error) ─► Линейный штраф. Устойчива к редким аномалиям
│   │
│   ├── На основе Среднего Геометрического ошибок
│   │   └── MSLE (Mean Squared Logarithmic Error) ──► Оценивает относительную ошибку в процентах
│   │
│   └── Гибридные и гладкие функции
│       ├── Huber Loss ──► Гибрид. Возле нуля работает как MSE, на крупных ошибках — как MAE
│       └── Log-Cosh Loss ─► Гладкий аналог MAE. Легко дифференцируется во всех точках
│
└── ЗАДАЧИ РАНЖИРОВАНИЯ (Оптимизируют порядок объектов в списках)
    │
    ├── 1. Поточечный подход (Pointwise Loss)
    │   └── Обычные MSE / Log Loss ──► Оценивают каждый документ отдельно вне списка. Модель не знает, кто выше, а кто ниже.
    │
    ├── 2. Попарный подход (Pairwise Loss)
    │   ├── RankNet Loss ────────────► Берет пары документов. Оптимизирует вероятность того, что более полезный документ получит балл выше, чем менее полезный.
    │   └── LambdaRank Loss ─────────► Развитие RankNet. При подсчете градиента для пары документов учитывает, насколько сильно их перестановка изменит финальную метрику NDCG.
    │
    └── 3. Списочный подход (Listwise Loss)
        ├── LambdaMART Loss ─────────► Стандарт индустрии в бустингах (XGBoost/LightGBM). Оптимизирует структуру всего дерева решений напрямую под финальный список документов.
        ├── ListNet Loss ────────────► Превращает оценки релевантности всего списка в распределение вероятностей (через Softmax) и минимизирует кросс-энтропию между идеальным и предсказанным списком.
        └── ApproxNDCG Loss ─────────► Суррогатная функция. Математически сглаживает операцию сортировки в NDCG, позволяя нейросетям обучаться напрямую на приближенную метрику NDCG.
```
