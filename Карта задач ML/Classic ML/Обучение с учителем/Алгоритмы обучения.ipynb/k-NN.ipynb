1. Ему нужна размеченная выборка (Учитель)

Чтобы k-NN мог классифицировать новый объект, ему абсолютно необходима **база данных с уже готовыми правильными ответами**.

- **Пример:** Если мы хотим определять, что на фото — кот или собака, в памяти k-NN должна лежать таблица признаков, где напротив каждой строчки строго написано: `[Признаки] -> Кот` или `[Признаки] -> Собака`.
- Без этих «подсказок учителя» алгоритм просто не сможет принять решение.

2. Он решает задачи классификации и регрессии

Это две канонические задачи обучения с учителем:

- **В задачах классификации:** Алгоритм смотрит на k ближайших соседей нового объекта и отдает ему тот класс, который встречается **чаще всего** (голосование большинством).
- **В задачах регрессии:** Алгоритм берет числовые значения k ближайших соседей и находит их **среднее арифметическое** (или среднее взвешенное).

---

🧠 В чем его уникальность? (Lazy Learning)

В машинном обучении k-NN называют **«ленивым алгоритмом» (Lazy Learner)**.

- Обычные модели (нейросети, линейная регрессия) во время обучения долго изучают данные, строят формулу, а исходные данные потом выбрасывают. При предсказании они используют только готовую формулу.
- k-NN во время «обучения» **не делает ничего**, кроме сохранения данных в память. Вся работа (поиск соседей и расчет расстояний) происходит в момент предсказания.
