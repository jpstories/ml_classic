1. Ему нужна размеченная выборка (Учитель)

Чтобы k-NN мог классифицировать новый объект, ему абсолютно необходима **база данных с уже готовыми правильными ответами**.

- **Пример:** Если мы хотим определять, что на фото — кот или собака, в памяти k-NN должна лежать таблица признаков, где напротив каждой строчки строго написано: `[Признаки] -> Кот` или `[Признаки] -> Собака`.
- Без этих «подсказок учителя» алгоритм просто не сможет принять решение.

2. Он решает задачи классификации и регрессии

Это две канонические задачи обучения с учителем:

- **В задачах классификации:** Алгоритм смотрит на k ближайших соседей нового объекта и отдает ему тот класс, который встречается **чаще всего** (голосование большинством).
- **В задачах регрессии:** Алгоритм берет числовые значения k ближайших соседей и находит их **среднее арифметическое** (или среднее взвешенное).

---

🧠 В чем его уникальность? (Lazy Learning)

В машинном обучении k-NN называют **«ленивым алгоритмом» (Lazy Learner)**.

- Обычные модели (нейросети, линейная регрессия) во время обучения долго изучают данные, строят формулу, а исходные данные потом выбрасывают. При предсказании они используют только готовую формулу.
- k-NN во время «обучения» **не делает ничего**, кроме сохранения данных в память. Вся работа (поиск соседей и расчет расстояний) происходит в момент предсказания.

❓ 1. «Что произойдет с kNN, если мы увеличим количество признаков (фичей) до 1000? Поможет ли увеличение базы данных?» 

- **Ожидаемый ответ:** Наступит **«проклятие размерности»** (Curse of Dimensionality). В пространстве высокой размерности (1000D) объём пространства растет экспоненциально. Все точки разлетаются друг от друга и становятся примерно _равноудаленными_. Понятие «ближайший сосед» теряет смысл. Увеличение базы данных не спасет.
- **Как решить проблему (что сказать):** Нужно применить методы снижения размерности (PCA, UMAP, t-SNE) или перейти с Евклидова расстояния на Манхэттенское, которое в высокой размерности ведет себя стабильнее.

❓ 2. «Почему kNN называют "ленивым" алгоритмом (Lazy Learner)? Какова его вычислительная сложность?»

- **Ожидаемый ответ:** На этапе обучения (`.fit()`) kNN не делает **ничего** — он просто сохраняет обучающую выборку в память. Сложность обучения равна \(O(1)\).
- Вся тяжелая работа происходит на этапе предсказания (`.predict()`), когда для одной новой точки нужно посчитать расстояния до всего миллиона точек в базе. Базовая сложность предсказания (Brute-Force) — \(O(N \cdot D)\), где \(N\) — число строк, а \(D\) — число признаков.
- **Доп. вопрос:** _«Как это оптимизировать в продакшене?»_ — Отвечайте: использовать KD-деревья, Ball-деревья или библиотеки для приближенного поиска ближайших соседей (ANN), например, `Faiss` от Meta или `HNSW` (Иерархический ориентированный граф), который активно применяется в поиске Яндекса.

❓ 3. «Что произойдет с kNN, если один признак измеряется в миллионах (бюджет), а второй — в единицах (возраст)? Как это исправить?»

- **Ожидаемый ответ:** Признак с огромным масштабом полностью «заглушит» мелкий признак при возведении разностей в квадрат. Из-за этого kNN будет принимать решения только на основе бюджета, игнорируя возраст.
- **Решение:** Обязательная **нормализация (MinMaxScaler)** или **стандартизация (StandardScaler)** данных перед обучением. 

❓ 4. «Как изменится Bias (смещение) и Variance (разброс) модели kNN при изменении параметра k от 1 до бесконечности?» 

Это классический вопрос на понимание концепции Bias-Variance Tradeoff:

- **При \(k = 1\):** Модель слишком сложная, идеально подстраивается под каждую точку тренировочных данных. **Bias — низкий** (ошибка на обучении нулевая), но **Variance — высокий** (модель переобучена, любая шумная точка испортит предсказание).
- **При \(k \to N\) (растет):** Модель сглаживается, усредняет всё подряд. **Variance — низкий** (модель стабильна), но **Bias — высокий** (модель недообучена, она просто выдает самый частый класс во всей базе данных).

❓ 5. «Что делать, если в выборке сильный дисбаланс классов (например, 99% — легитимные транзакции, 1% — мошенничество)?»

- **Ожидаемый ответ:** kNN сломается, так как при голосовании мажоритарный класс всегда будет побеждать числом.
- **Решение:**
    1. Переключить kNN в режим **взвешенного голосования** (`weights='distance'`) — тогда близкий сосед-мошенник получит огромный вес и перевесит толпу далеких обычных клиентов.
    2. Применить алгоритмы ресемплинга выборки (например, SMOTE, чтобы искусственно размножить редкий класс).
