## 1. Метрики классификации (Предсказание класса)

В основе оценки классификаторов лежит **Матрица ошибок (Confusion Matrix)**. Она делит все ответы модели на 4 исхода.

Представьте, что мы ищем бракованные детали на заводе:

- **TP (True Positive):** Модель сказала «брак», и деталь реально бракованная (Верное попадание).
    
- **TN (True Negative):** Модель сказала «норма», и деталь нормальная (Успешный пропуск).
    
- **FP (False Positive):** Модель сказала «брак», а деталь идеальная (Ложная тревога / Ошибка I рода).
    
- **FN (False Negative):** Модель сказала «норма», а деталь бракованная (Пропуск цели / Ошибка II рода).
    

### Accuracy (Доля правильных ответов)

Самая интуитивная метрика, но невероятно опасная в реальной жизни.

- **Логика:** Какую долю объектов модель угадала верно?
    
- **Почему опасна:** Представьте, что на заводе только 1% брака. Если модель просто сломается и будет отвечать «норма» на 100% деталей, её Accuracy составит 99%. Цифра красивая, но модель абсолютно бесполезна. Accuracy можно использовать только если классы равны (50% на 50%).
    

### Precision (Точность)

- **Логика:** «Если модель сказала "Да", насколько ей можно верить?»
    
    $$Precision=\frac{TP}{TP+FP}$$
    
- **Пример в бизнесе:** Спам-фильтр почты. Здесь ложная тревога ($FP$) стоит дорого — алгоритм удалит важное письмо от клиента. Нам лучше пропустить немного спама в инбокс, чем потерять сделку. Задача — выкрутить Precision на максимум.
    

### Recall (Полнота)

- **Логика:** «Из всех реальных "Да", сколько штук модель смогла найти?»
    
    $$Recall=\frac{TP}{TP+FN}$$
    
- **Пример в бизнесе:** Диагностика онкологии. Здесь ложная тревога ($FP$) стоит копейки — врач просто назначит пациенту еще один анализ. А вот пропуск больного ($FN$) стоит жизни. Задачей становится поиск всех больных любой ценой (максимальный Recall).
    

### F1-Score

- **Логика:** Баланс между Precision и Recall (их гармоническое среднее).
    
- **Когда использовать:** Когда вам важны и ложные срабатывания, и пропуски, но классы сильно не сбалансированы. F1-Score упадет близко к нулю, если хотя бы одна из метрик (Precision или Recall) просядет.

## 2. Метрики регрессии (Предсказание числа)

Когда мы предсказываем конкретные числа (цену квартиры, время в пути, продажи), понятие «угадал/не угадал» не работает. Модель не угадает цену с точностью до копейки. Здесь мы измеряем силу промаха.

### MAE (Средняя абсолютная ошибка)

- **Логика:** Средняя ошибка в тех же единицах измерения, что и целевая переменная.
    
    $$MAE=\frac{1}{n}\sum_{i=1}^{n}\vert{}y_i-\hat{y}_i\vert{}$$
    
- **Плюс:** Идеальна для объяснения бизнесу. «Наша модель ошибается при оценке квартиры в среднем на 300 000 рублей».
    
- **Минус:** Одинаково относится к маленьким и большим ошибкам.
    

### MSE (Среднеквадратичная ошибка)

- **Логика:** Возводит каждую ошибку в квадрат.
    
    $$MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$$
    
- **Зачем нужна:** Квадрат заставляет модель панически бояться **больших промахов**. Ошибиться в прогнозе поставок на 10 тонн — это не в 10 раз хуже, чем ошибиться на 1 тонну. Зачастую это в 100 раз хуже из-за штрафов от складов и порчи товара. MSE наказывает модель за такие выбросы.
    
- **Минус:** Измеряется в «квадратных рублях» или «квадратных тоннах», что невозможно объяснить заказчику.
    

### RMSE (Корень из среднеквадратичной ошибки)

- **Логика:** Корень из $MSE$.
    
- **Зачем нужна:** Возвращает ошибку обратно в нормальные единицы измерения (в рубли, а не квадратные рубли), при этом сохраняя строгое наказание за большие выбросы. Является стандартом по умолчанию для 90% задач регрессии.
    

### R² (Коэффициент детерминации)

- **Логика:** Показывает, насколько наша модель справляется лучше, чем если бы мы просто брали среднее историческое значение. Измеряется от 0 до 1 (или в процентах).
    
- **Интерпретация:** $R^2=0.85$ означает, что наша модель учитывает 85% всех факторов, влияющих на целевую переменную. Если $R^2<0$, это катастрофа: модель работает хуже, чем просто среднее арифметическое.
    

## Сводная таблица для бизнеса

|**Задача**|**Опасность**|**Какую метрику максимизировать / минимизировать**|
|---|---|---|
|Спам-фильтр, Выдача банов|Ошибочно наказать хорошего пользователя ($FP$)|Максимизировать **Precision**|
|Медицина, Антифрод, Техобслуживание|Пропустить аварию или болезнь ($FN$)|Максимизировать **Recall**|
|Кредитный скоринг|Нужен компромисс между рисками и прибылью|Максимизировать **F1** или **ROC-AUC**|
|Прогноз стабильных продаж|Бизнесу нужна понятная ошибка в рублях/штуках|Минимизировать **MAE**|
|Прогноз спроса на скоропорт|Редкие сильные ошибки ведут к порче тонн еды|Минимизировать **RMSE**|

# 3. ROC-AUC: Золотой стандарт классификации

Алгоритмы машинного обучения (например, логистическая регрессия или градиентный бустинг) редко выдают жесткий ответ «Да» или «Нет». Обычно они выдают **вероятность** — например, «этот клиент вернет кредит с вероятностью 0.85».

Чтобы перевести вероятность в финальное решение (выдать кредит или отказать), нам нужен **порог отсечения (threshold)**. По умолчанию он равен 0.5. Но если мы изменим порог на 0.8, метрики _Precision_ и _Recall_ кардинально изменятся.

**Главная проблема:** Как оценить качество самой модели, не привязываясь к конкретному порогу? Именно эту задачу решает ROC-AUC.

## 1. Как читать график (ROC-кривая)

**ROC-кривая (Receiver Operating Characteristic)** — это график, который показывает, как меняется качество модели при переборе всех возможных порогов отсечения (от 0.0 до 1.0).

У графика есть две оси:

- **Ось Y (TPR - True Positive Rate):** Это тот же самый _Recall_ (Полнота). Доля найденных реальных «единиц». Мы хотим, чтобы она стремилась к $100\%$ ($1.0$).
    
    $$TPR = \frac{TP}{TP + FN}$$
- **Ось X (FPR - False Positive Rate):** Доля ложных тревог. Процент объектов класса «ноль», которые модель ошибочно назвала «единицей». Мы хотим, чтобы она стремилась к $0\%$ ($0.0$).
    
    $$FPR = \frac{FP}{FP + TN}$$

**Как строится кривая:**

1. Мы ставим порог $1.0$ (никто не получает класс «1»). Модель не делает ложных тревог ($FPR = 0$), но и никого не находит ($TPR = 0$). Это точка $(0, 0)$ в левом нижнем углу.
    
2. Мы постепенно снижаем порог. Начинают появляться как верные попадания, так и ложные тревоги. Линия ползет вверх и вправо.
    
3. При пороге $0.0$ всем объектам присваивается класс «1». Модель нашла всех ($TPR = 1$), но подняла 100% ложных тревог ($FPR = 1$). Это точка $(1, 1)$ в правом верхнем углу.
    

**Диагональ графика** (из левого нижнего в правый верхний угол) — это случайное угадывание (бросок монетки). Хорошая модель всегда должна выгибаться дугой **выше** этой диагонали, стремясь в идеальный левый верхний угол (где $TPR=1$, а $FPR=0$).

## 2. Что такое AUC?

Сама по себе кривая — это просто картинка. Чтобы сравнивать модели программно, нужно одно число.

**AUC (Area Under Curve)** — это площадь под ROC-кривой.

- $AUC = 0.5$ (Площадь под диагональю) — Модель абсолютно бесполезна, она гадает наугад.
    
- $AUC = 1.0$ (Вся площадь квадрата) — Идеальная модель, которая безошибочно разделяет классы.
    
- $AUC \approx 0.7 - 0.8$ — Нормальная, рабочая модель.
    
- $AUC \approx 0.8 - 0.9$ — Отличная модель.
    
- $AUC < 0.5$ — Модель работает "наоборот" (ошибается чаще, чем гадает). Если поменять её ответы на противоположные, она станет хорошей.
    

## 3. Магическая интерпретация AUC (Почему это золотой стандарт?)

У площади под кривой (AUC) есть гениальное математическое свойство, которое делает её лучшей метрикой для бизнеса:

> **AUC — это вероятность того, что если мы возьмем один случайный "хороший" объект и один случайный "плохой", модель даст "хорошему" объекту балл выше, чем "плохому".**

**Пример из жизни:** Представьте, вы разработали систему кредитного скоринга с $AUC = 0.85$. Вы берете из истории банка одного случайного дефолтника (класс 0) и одного случайного человека, вернувшего долг (класс 1). С вероятностью $85\%$ ваша модель присвоила честному заемщику скоринговый балл **выше**, чем дефолтнику.

### Почему бизнес любит ROC-AUC:

1. **Оценка ранжирования, а не угадывания.** Модели часто используют для сортировки (например, кому из клиентов позвонить первому, чтобы предложить услугу). ROC-AUC идеально показывает, насколько хорошо модель сортирует клиентов от наиболее вероятных к наименее вероятным.
    
2. **Игнорирование дисбаланса.** В отличие от _Accuracy_, ROC-AUC не ломается, если у вас 99% нулей и 1% единиц.
    
3. **Разделение труда.** Дата-саентист может обучить модель с высоким AUC (обеспечить хорошее разделение классов), а бизнес-заказчик потом сам выберет нужный порог отсечения (Threshold) на основе своих финансовых рисков, двигаясь вдоль ROC-кривой.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, roc_auc_score

# 1. Генерируем синтетические данные (например, профили клиентов для кредита)
# 1000 объектов, 10 признаков, из которых 5 действительно важны для предсказания
X, y = make_classification(n_samples=1000, n_features=10, 
                           n_informative=5, random_state=42)

# 2. Разбиваем данные на обучающую (70%) и тестовую (30%) выборки
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Инициализируем и обучаем модель
model = LogisticRegression()
model.fit(X_train, y_train)

# 4. Получаем ВЕРОЯТНОСТИ принадлежности к классу 1 (а не просто 0 или 1)
# Метод predict_proba возвращает вероятности для обоих классов: [вероятность_0, вероятность_1]
# Нас интересует только столбец с индексом 1 (вероятность целевого события)
y_scores = model.predict_proba(X_test)[:, 1]

# 5. Считаем метрики для ROC-кривой
# fpr - False Positive Rate (Доля ложных тревог по оси X)
# tpr - True Positive Rate (Полнота по оси Y)
# thresholds - Пороги отсечения, при которых получились эти точки
fpr, tpr, thresholds = roc_curve(y_test, y_scores)

# Считаем саму площадь под кривой (AUC)
roc_auc = roc_auc_score(y_test, y_scores)

# 6. Отрисовываем график
plt.figure(figsize=(8, 6))

# Рисуем ROC-кривую нашей модели
plt.plot(fpr, tpr, color='darkorange', lw=2, 
         label=f'ROC-кривая модели (AUC = {roc_auc:.3f})')

# Рисуем пунктирную линию случайного угадывания (AUC = 0.5)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', 
         label='Случайное угадывание (AUC = 0.500)')

# Наводим красоту: подписи осей, заголовок, сетка
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (Доля ложных тревог)')
plt.ylabel('True Positive Rate (Полнота / Доля найденных)')
plt.title('ROC-кривая Логистической регрессии')
plt.legend(loc="lower right")
plt.grid(alpha=0.3)

# Показываем результат
plt.show()
```
