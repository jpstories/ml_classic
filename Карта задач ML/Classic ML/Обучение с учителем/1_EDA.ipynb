==**Exploratory Data Analysis**== (разведочный или исследовательский анализ данных). Это предварительный этап работы с информацией, на котором специалист изучает структуру, основные свойства, аномалии и взаимосвязи в данных с помощью визуализации и статистики до обучения модели.

Решает

- **Поиск ошибок:** обнаружение пропущенных значений (null/NaN) и дубликатов.

- **Поиск аномалий:** выявление резких выбросов, способных ухудшить качество модели.

- **Оценка связей:** анализ корреляции между признаками и целевой переменной.

- **Баланс классов:** проверка набора данных на перекосы (например, в задачах классификации).

Основные методы и инструменты

- **Визуализация:** построение графиков распределения, ящиков с усами (box-plot) и матриц корреляции.

- **Статистика:** расчет средних значений, медиан, стандартных отклонений и квантилей.

- **Библиотеки Python:** использование инструментов `pandas`, `numpy`, `matplotlib` и `seaborn`.

Шаг 1. Первое знакомство с данными

Сначала оцениваем размер таблицы, типы данных и смотрим на первые строки.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Создаем данные прямо в коде, чтобы не зависеть от интернета и файлов
data = {
    'User_ID': [1,2,3,4,5],
    'Age': [30, 21, 35, 42, 19],
    'Salary': [50000, 120000, 85000, 50000, None],  # Добавили один пропуск (None)
    'Purchased': ['No', 'Yes', 'Yes', 'No', 'Yes']
}

# Превращаем словарь в DataFrame Pandas
df = pd.DataFrame(data)

# Просмотр первых 5 строк
print("--- ПЕРВЫЕ 5 СТРОК ДАТАСЕТА ---")
print(df.head())

print("\n--- ИНФОРМАЦИЯ О ТИПАХ ДАННЫХ ---")
# Информация о типах данных и количестве заполненных строк
print(df.info())
```

Шаг 2. Анализ пропусков и дубликатов

Пропуски могут сломать алгоритмы машинного обучения, а дубликаты смещают метрики.

```python
# Поиск дубликатов
print(f"Дубликатов: {df.duplicated().sum()}")

# Подсчет пропущенных значений в каждом столбце (в процентах)
missing_percent = df.isnull().mean() * 100
print(missing_percent[missing_percent > 0])
```

Шаг 3. Базовая статистика и поиск выбросов

Изучаем масштабы данных, средние значения и разброс. Выбросы ищем с помощью графика **Boxplot** («ящик с усами»).

```python
# Описательная статистика для числовых признаков
print(df.describe())

# Визуализация выбросов для конкретного признака (например, Price)
sns.boxplot(x=df['Price'])
plt.title('Поиск выбросов в ценах')
plt.show()
```

Шаг 4. Распределение признаков

Проверяем форму распределения данных. Для этого строим гистограммы.

```python
# График распределения для числового признака (например, Age)
sns.histplot(df['Age'], kde=True)
plt.title('Распределение возраста')
plt.show()

# Для категориальных признаков смотрим баланс классов
sns.countplot(x=df['Target_Class'])
plt.title('Баланс целевых классов')
plt.show()
```

Шаг 5. Поиск взаимосвязей и корреляции

Изучаем, как признаки связаны между собой и с целевой переменной. Линейные связи показывает тепловая карта (**Heatmap**).

```python
# Расчет матрицы корреляции (только для числовых столбцов)
corr_matrix = df.select_dtypes(include=['number']).corr()

# Строим тепловую карту
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Матрица корреляции признаков')
plt.show()
```
