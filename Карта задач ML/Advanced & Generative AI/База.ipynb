Классическое машинное обучение (Деревья, Линейные модели)

👍 Плюсы:

- **Скорость обучения:** Модели вроде XGBoost или Random Forest обучаются на процессоре обычного ноутбука за секунды или минуты.
- **Мало данных? Не проблема:** Классический ML отлично работает, если у вас в таблице всего 500 или 1000 строк.
- **Интерпретируемость («Белый ящик»):** Вы всегда можете открыть дерево решений и посмотреть, почему оно приняло такое решение (например, _«Станок сломался, потому что вибрация > 3.5»_). Это критично в медицине, банках и на опасных производствах.
- **Простота внедрения:** Обученную модель можно сохранить в крошечный файл (пару килобайт) и запустить даже на слабом контроллере внутри датчика.

👎 Минусы:

- **Ручная работа (Feature Engineering):** Алгоритмы глупы сами по себе. Если вы не посчитаете скользящее окно, не выделите день недели из даты или не очистите данные от мусора вручную, модель выдаст плохой результат.
- **Потолок точности:** Сколько бы данных вы ни вливали в Случайный лес, после определенного объема (например, 1 миллион строк) его точность перестает расти. Он упирается в свой математический потолок.

Нейронные сети (Глубокое обучение)

👍 Плюсы:

- **Автоматический поиск признаков:** Вам не нужно выдумывать хитрые скользящие окна или тренды. Вы можете «вывалить» в сеть терабайты сырых данных, и за счет множества слоев она сама найдет скрытые нелинейные закономерности.
- **Нет потолка точности:** Чем больше данных и чем больше нейросеть, тем точнее она становится. Она буквально «кушает» данные и постоянно улучшается.
- **Работа со сложными типами данных:** Классический ML беспомощен, если ему дать картинку (матрицу пикселей), видео, живой человеческий голос или текст. Нейросети (сверточные, трансформеры) щелкают эти задачи как орехи.

👎 Минусы (То, о чём вы сказали):

- **Долгое и дорогое обучение:** Вместо секунд нейросети обучаются часами, днями или неделями. Для этого нужны дорогие видеокарты (GPU). Ошибка в коде может стоить тысяч долларов, сожженных на аренду серверов.
- **Сложность использования (Deployment):** Обученная нейросеть весит гигабайты. Чтобы она выдавала предсказания на заводе или в приложении, рядом нужно ставить мощный сервер с видеокартой или использовать облако.
- **«Черный ящик»:** Вы никогда не узнаете, почему нейросеть выдала именно такой ответ. Внутри нее миллиарды переплетенных коэффициентов. Если она ошибется, расследовать причину будет невероятно сложно.
- **Капризность к масштабу данных:** Как мы разбирали в самом начале, забыли нормализовать данные — сеть «ослепла» и выдает случайный шум. Классическим деревьям на это плевать.

---

Золотое правило инженера (Что выбирать?)

1. Если ваши данные — это **таблицы** (базы данных, логи, показатели датчиков), а строк у вас меньше пары миллионов \(\rightarrow \) **используйте классическое ML** (XGBoost, CatBoost, LightGBM). Это быстрее, дешевле и зачастую точнее.
2. Если ваши данные — это **картинки, видео, звук, тексты** или колоссальные массивы сложных сигналов, где классические деревья уже не справляются \(\rightarrow \) **идите в нейросети**.
