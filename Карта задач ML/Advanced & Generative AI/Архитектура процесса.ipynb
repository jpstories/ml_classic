```python
# Исходные данные
x = 2.0
y = 10.0
w = 3.0  # Стартовый вес
learning_rate = 0.1

# 1. Прямой ход (Forward pass) — сеть делает предсказание
y_pred = w * x  # 3.0 * 2.0 = 6.0

# 2. Считаем производную по нашей формуле (Градиент)
# На сколько нужно изменить вес?
dw = 2 * (y_pred - y) * x  # 2 * (6.0 - 10.0) * 2.0 = -16.0

# 3. Обновляем вес (Шаг градиентного спуска)
# w = 3.0 - (0.1 * -16.0) -> w = 3.0 + 1.6 -> w = 4.6
w = w - learning_rate * dw

print(f"Новый вес: {w}") 
# Можете проверить: новое предсказание будет 4.6 * 2 = 9.2. 
# Ошибка стала (9.2 - 10)^2 = 0.64. А была 16! Производная сработала идеально.
```

Вместо того чтобы писать отдельную строчку для каждого входа, программисты объединяют все входы в один вектор (массив X), а веса — в вектор W.   
Перемножаются они одной быстрой командой (матричным умножением).

```python
import numpy as np

# Входы (площадь=50, метро=10, этаж=5)
X = np.array([50.0, 10.0, 5.0])
y = 5000000.0  # Реальная цена квартиры

# Начальные случайные веса для каждого входа
W = np.array([20000.0, -5000.0, 1000.0])
learning_rate = 0.0001

# 1. Прямой ход (перемножаем входы на веса и складываем)
y_pred = np.sum(X * W)

# 2. Считаем ошибку
loss = (y_pred - y) ** 2

# 3. Считаем производные (градиент) сразу для ВСЕХ весов!
# NumPy сделает это одной строчкой для каждого элемента массива
dW = 2 * (y_pred - y) * X

# 4. Обновляем все веса одновременно
W = W - learning_rate * dW
```

Когда на вход нейросети подается картинка (например, 28 х 28 пикселей), у неё на самом деле просто 784 входа (каждый пиксель — отдельное число). Но код под капотом делает ровно то же самое: считает 784 производные и одновременно их обновляет.
