#### 1. Используйте Torchviz, когда: **«Я пишу нестандартную сеть, и она выдает ошибку размерностей или градиентов»**

Torchviz не просто рисует слои (как квадратики), он рисует **путь распространения ошибки** (backpropagation).

- **Идеально для:** Сложных архитектур с разветвлениями (например, ResNet со skip-connections) или нейросетей с кастомными функциями потерь.
    
- **Решает проблему:** Если при обучении градиент «затухает» или отваливается (возникает ошибка _RuntimeError: element 0 of tensors does not require grad_), Torchviz покажет, в каком именно месте прервалась связь между слоями.
    

#### 2. Используйте TensorBoard, когда: **«Я запустил обучение и хочу знать, учится ли модель вообще»**

Это базовый инструмент мониторинга. Никто в индустрии не смотрит на бегущие строчки в терминале, все смотрят на графики.

- **Идеально для:** Отслеживания переобучения (overfitting). Вы выводите на один график Loss на тренировочной выборке и Loss на валидационной. Если тренировочный падает, а валидационный начал расти — модель переобучилась.
    
- **Решает проблему:** Позволяет визуально сравнивать разные эксперименты (например, обучение с learning rate 0.01 против 0.001) и смотреть на гистограммы распределения весов (чтобы убедиться, что веса не превратились в нули или бесконечности).
    

#### 3. Используйте Netron, когда: **«Мне нужно понять, как устроена готовая модель, или подготовить её к внедрению»**

Netron показывает нейросеть такой, какой её видит «железо» при выполнении логического вывода (инференса).

- **Идеально для:** Случаев, когда вы скачали чужую модель (например, YOLO для детекции объектов) и вам нужно точно узнать, картинку какого размера она принимает на вход (`[1, 3, 224, 224]`) и в каком формате выдает результат.
    
- **Решает проблему:** Дебаггинг перед продакшеном. Перед тем как отдать модель бэкенд-разработчикам (или загрузить на мобильное устройство), ML-инженер конвертирует её в формат ONNX и проверяет в Netron, нет ли внутри лишних операций, которые замедлят работу.
