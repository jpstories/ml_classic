## Моя оценка по компонентам

### 1. Академическая база — 10/10

Это огромный плюс.

Большинство ML-инженеров умеют пользоваться PyTorch.

Очень немногие понимают:

- почему LayerNorm работает;
- почему Adam стабилен;
- почему появляется gradient explosion;
- что именно делает backpropagation.

Через несколько лет именно фундамент будет отличать инженеров, которые могут создавать новые решения, от тех, кто умеет только собирать пайплайны.

---

### 2. Стандарт качества кода — 10/10

Это даже лучше, чем у многих ML-команд.

Type hints.

Документация.

Docker.

Тесты.

CI.

Линтеры.

Именно так выглядит инженерия.

---

### 3. Системное мышление — 9.5/10

Очень хороший раздел.

Но я бы расширил его до понятия **ML Systems**.

То есть добавить понимание:

- GPU;
- память;
- latency;
- throughput;
- batching;
- profiling;
- distributed training.

Именно здесь сегодня происходит большая часть инженерной работы.

---

### 4. MLOps — 9/10

Очень достойно.

Но не стоит превращать изучение MLOps в самоцель.

Для Applied ML Engineer это инструмент, а не основная специализация.

---

### 5. Robotics — 9.5/10

Очень правильная долгосрочная цель.

Особенно потому, что сейчас развивается направление Embodied AI.

Но роботика имеет смысл только после действительно сильной базы в DL и ML Systems.

## Что делать самому

Допустим, вы пишете Transformer.

Я бы сказал:

**Не открывайте ChatGPT.**

Сначала самостоятельно:

- спроектируйте классы;
- напишите MultiHeadAttention;
- реализуйте LayerNorm;
- реализуйте FeedForward;
- реализуйте Positional Encoding;
- реализуйте Masking;
- реализуйте обучение.

Пусть даже получится плохо.

Пусть будут ошибки.

Это нормально.

Именно в этот момент развивается инженерное мышление.

## Когда подключать ИИ

После этого можно спросить:

> "Посмотри мой MultiHeadAttention. Где узкие места?"

или

> "Почему здесь растёт VRAM?"

или

> "Как можно ускорить этот DataLoader?"

или

> "Я не понимаю, почему градиенты стали NaN."

Это очень похоже на взаимодействие с опытным коллегой во время code review.

# Что категорически нельзя делать

Самая большая ошибка выглядит так:

> "Напиши мне Transformer."

Потом:

копировать

вставить

запустить

готово.

Через неделю вы уже не сможете объяснить:

- зачем нужен residual;
- зачем LayerNorm;
- почему attention масштабируется на √d;
- как считается backprop.

Такой подход создаёт иллюзию знаний.

# Как работают сильные инженеры

Представим инженера из NVIDIA.

Он не пишет весь код вручную.

Он использует Copilot, ChatGPT и другие инструменты.

Но есть принципиальная разница.

Он **может написать этот код сам**.

ИИ лишь ускоряет работу.

# Моя схема обучения

### Шаг 1

Получили задачу.

Например:

> Реализовать CNN.

---

### Шаг 2

Без ИИ.

Продумываете архитектуру.

Пишете код.

---

### Шаг 3

Запускаете.

Получаете 20 ошибок.

Исправляете.

---

### Шаг 4

Если застряли больше чем на 30–60 минут, обращаетесь к ИИ.

Например:

> "Не получается понять, почему размерности после Conv2D не совпадают. Не пиши решение, помоги найти ошибку."

Это гораздо полезнее, чем получить готовый ответ.

---

### Шаг 5

Когда проект готов:

просите ИИ провести полноценное ревью:

- архитектуры;
- читаемости;
- производительности;
- стиля;
- тестов;
- потенциальных улучшений.

# Почему я так настаиваю

Потому что на собеседовании будет примерно так.

Интервьюер:

> Почему здесь используется LayerNorm?

Если код писал ИИ,

ответ обычно такой:

> "Ну... так принято..."

Если код писали вы,

ответ звучит иначе:

> "Я сначала попробовал без LayerNorm. Обучение стало нестабильным примерно после N эпох. Потом реализовал LayerNorm и увидел, что распределение активаций стабилизировалось..."

Это совершенно другой уровень понимания.

# Правило, которое я бы сделал обязательным на все два года

> **Если вы не можете реализовать ключевой алгоритм без помощи ИИ, значит, вы его ещё не освоили.**

А вот после того как вы **можете** написать его самостоятельно, использование ИИ становится огромным преимуществом. Он начинает экономить время на рутинных задачах, помогает находить ошибки и улучшать код, не подменяя собой ваше понимание.

Именно так я бы рекомендовал строить обучение человеку, который хочет стать инженером мирового уровня, а не просто научиться пользоваться современными AI-инструментами.
