# Что важно: не выбрасывать твой full stack опыт

Это очень частая ошибка.

Люди думают:

> "Я был full stack, значит это прошлое."

Нет.

В AI-продуктах очень ценятся люди, которые могут сделать:

```
модель

↓

backend

↓

API

↓

frontend

↓

production
```

Исследователь может придумать модель.

ML Engineer может обучить.

Но кто доведет продукт до пользователей?

Таких людей меньше.
