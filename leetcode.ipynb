1. Тема «Два указателя» и «Скользящее окно» (Абсолютный топ в Яндексе)

_Эти задачи проверяют умение оптимизировать алгоритмы со сложности O(N²) до O(N) по времени._

- **1** — Two Sum _(Easy)_ — база, с которой все начинается.
- **283** — Move Zeroes _(Easy)_ — классическая разминочная задача на указатели.
- **392** — Is Subsequence _(Easy)_ — часто дают в самом начале интервью.
- **11** — Container With Most Water _(Medium)_ — отличная задача на схождение указателей с краев.
- **15** — 3Sum _(Medium)_ — усложненная вариация базовой задачи, проверяет обработку дубликатов.
- **3** — Longest Substring Without Repeating Characters _(Medium)_ — классика скользящего окна.
- **438** — Find All Anagrams in a String _(Medium)_ — окно фиксированного размера + хеш-таблица.

2. Тема «Массивы, Строки и Хеш-таблицы»

_Проверка базовой работы с памятью, строками и эффективным поиском за O(1)._

- **121** — Best Time to Buy and Sell Stock _(Easy)_ — линейный обход массива.
- **49** — Group Anagrams _(Medium)_ — группировка данных с использованием хеш-мапы как ключа.
- **238** — Product of Array Except Self _(Medium)_ — задача на префиксные/суффиксные произведения без использования деления.
- **347** — Top K Frequent Elements _(Medium)_ — подсчет частот + сортировка/куча.
- **560** — Subarray Sum Equals K _(Medium)_ — комбинация префиксных сумм и хеш-таблицы (очень любят в Яндексе).

3. Тема «Объединение интервалов» (Специфика Яндекса)

_Задачи на геометрию отрезков. В Яндексе их дают регулярно (например, задачи про расписание встреч, свободные слоты в календаре или доставку)._

- **56** — Merge Intervals _(Medium)_ — базовая и самая важная задача в этой теме.
- **57** — Insert Interval _(Medium)_ — вставка нового отрезка в уже отсортированный список.
- **986** — Interval List Intersections _(Medium)_ — поиск пересечений двух списков интервалов.

4. Тема «Обходы графов и матриц (BFS / DFS)»

_Поскольку вы идете в Computer Vision, понимание работы с двумерными матрицами как с графами (пиксели, маски сегментации) проверяется обязательно._

- **200** — Number of Islands _(Medium)_ — поиск связанных компонент в матрице (классический DFS/BFS).
- **695** — Max Area of Island _(Medium)_ — усложненная версия предыдущей задачи с подсчетом площади.
- **542** — 01 Matrix _(Medium)_ — поиск кратчайшего расстояния во все стороны (идеально для BFS).
- **994** — Rotting Oranges _(Medium)_ — классическая задача на послойный BFS (моделирование распространения процесса).

5. Тема «Деревья и Связные списки»

_Проверка понимания работы со ссылками в памяти и рекурсией._

- **206** — Reverse Linked List _(Easy)_ — разворот списка (часто просят написать за 5 минут без багов).
- **141** — Linked List Cycle _(Easy)_ — алгоритм "черепахи и зайца" (указатели с разной скоростью).
- **102** — Binary Tree Level Order Traversal _(Medium)_ — обход дерева по уровням (BFS).
- **104** — Maximum Depth of Binary Tree _(Easy)_ — базовая рекурсия на дерево.
- **236** — Lowest Common Ancestor of a Binary Tree _(Medium)_ — поиск наименьшего общего предка (проверка глубокого понимания рекурсии).
- **98** — Validate Binary Search Tree _(Medium)_ — проверка того, является ли дерево деревом поиска.
