# Но я бы добавил еще один проект

И, честно говоря, **он мне нравится даже больше остальных.**

## Project Zero

Это не один проект.

Это библиотека, которую ты будешь писать **все пять лет**.

Например:

```
phoenix/

tensor/

autograd/

optimizers/

nn/

cnn/

transformer/

diffusion/

cuda/

renderer/

vision/

robotics/

benchmarks/

notes/
```

Каждый новый проект должен **не жить отдельно**, а добавлять в `phoenix` новый модуль.

Например:

Сегодня изучил свертки.

Добавил

```
Conv2D
```

Через месяц:

```
Attention
```

Через полгода:

```
FlashAttention
```

Потом:

```
CUDA backend
```

Через пять лет получится не "домашняя работа", а **собственный AI-фреймворк**, пусть и учебный.

# И вот идея, которая, как мне кажется, тебе подойдет идеально

Представь, что ты не студент.

Ты — **руководитель маленькой исследовательской лаборатории из одного человека**.

Каждые 2–3 месяца лаборатория берет новую задачу.

Например:

**Квартал 1**

> Сделать лучший AI-ремастер текстур для Morrowind.

**Квартал 2**

> Сделать генерацию анимаций.

**Квартал 3**

> Написать собственный CUDA kernel для Attention.

**Квартал 4**

> Повторить статью по Gaussian Splatting.

Каждый квартал заканчивается не экзаменом, а работающим прототипом и новым модулем в `phoenix`.

## И, наверное, это будет моим главным советом на ближайшие годы.

**Не пытайся стать "экспертом по AI".** Это слишком расплывчатая цель.

Стань человеком, который **регулярно доводит до конца сложные инженерные проекты**. Если делать это несколько лет подряд, глубокие знания появятся как следствие.

И если бы я выбирал только **один** проект, который лучше всего отражает твои интересы, я бы выбрал именно **AI Remaster**. Он удивительным образом объединяет почти всё, что тебя вдохновляет:

- машинное обучение;
- генеративные модели;
- компьютерное зрение;
- графику;
- GPU и оптимизацию;
- C++;
- и, что немаловажно, эмоциональную составляющую — ты буквально возвращаешь к жизни игры, которые тебе дороги.

Такой проект достаточно большой, чтобы расти вместе с тобой несколько лет: начать с готовых моделей, затем дообучать их, потом ускорять инференс, а позже заменить отдельные компоненты собственными реализациями. Это уже не просто учебное упражнение, а направление, вокруг которого можно выстроить значительную часть своего развития.
