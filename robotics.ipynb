### Карта софта и материалов для старта (Первый месяц)

Чтобы не утонуть в тоннах старых учебников по C++, используй только современные и практические ресурсы.

1. Язык C++ (Стандарты C++17/C++20)

Забудь про учебники из нулевых. Тебе нужен C++ как инструмент высокопроизводительного бэкенда.

- **Где учить:** Интерактивный бесплатный учебник **[LearnCpp.com](https://www.learncpp.com/)** — это лучший мировой стандарт. Иди сразу к главам про указатели (Pointers), ссылки (References), динамическое выделение памяти и Smart Pointers (умные указатели).
- **Где писать код:** Ставь **VS Code** или **CLion** (от JetBrains). Настрой компилятор GCC и систему сборки **CMake** — это стандарт для ROS 2.

2. Знакомство с ROS 2 (Robot Operating System)

Параллельно с C++ начни смотреть, как устроена архитектура робота.

- **Что читать:** Официальная документация **[ROS 2 Documentation (версия Jazzy или Humble)](https://docs.ros.org/)**. Раздел _Tutorials_ (CLI и Client Libraries).
- **Главная концепция для понимания:** Представь, что код робота — это микросервисная архитектура. Сервис камеры каждую секунду публикует массив байт (картинку) в сеть, а сервис ИИ-детектора подписан на этот поток, обрабатывает его и шлет команды моторам.

Твоя цель на этот год — вакансии с названиями **ML Engineer (MLE)** или **Computer Vision Engineer**.

- **Твое позиционирование на собеседованиях:** «Я сильный бэкенд-инженер (6 лет опыта), который умеет оптимизировать код, писать быстрые API и разворачивать микросервисы, но теперь я хочу автоматизировать деплой моделей и работать с видеопотоком/изображениями».

Им постоянно нужны люди, которые делают видеоаналитику для магазинов (детекция очередей, подсчет товаров на полках, распознавание лиц). Это на 100% софтверная работа на Python, но она напрямую готовит тебя к роботам.

Шаг 2. Какой стек качать в этот «промежуточный» год

Этот стек сделает тебя высокооплачиваемым MLE сейчас и закроет половину требований в робототехнике через год:

1. **FastAPI вместо Django:** Забудь про Django на этот год. Все ML-сервисы в продакшене пишутся на FastAPI. Он асинхронный, быстрый и легкий.
2. **PyTorch & OpenCV:** Твоя база для работы с картинками и видео. Научись брать предобученные нейросети (например, YOLOv8 или EfficientNet) и дообучать их под конкретные задачи (Transfer Learning).
3. **Оптимизация инференса (Киллер-фича):** Изучи **ONNX Runtime** и **NVIDIA TensorRT**. Робот на борту имеет слабое железо, и твоя задача как инженера — сделать так, чтобы сетка выдавала не 2 кадра в секунду, а 60. В коммерческом вебе за это тоже платят огромные деньги.
4. **Docker + Командная строка Linux:** Стань богом терминала Ubuntu. Роботы, как и современные ML-сервисы, разворачиваются в контейнерах.

Тема 1: Свёрточные нейронные сети (CNN) и классическая обработка изображений

Это база, благодаря которой код начинает «видеть». На собеседованиях тебя будут гонять по механике того, как картинка превращается в набор признаков.

- **Что нужно знать «на пальцах»:**
    - Как работает операция свёртки (Convolution) и пулинга (Pooling). Зачем они нужны (сжатие размерности, выделение паттернов: граней, углов).
    - Архитектуры, ставшие классикой: **ResNet** (главное — понять концепцию _Skip Connections_, которая решила проблему затухания градиентов) и **EfficientNet** (эталон по соотношению скорость/точность).
- **Практический навык:**
    - Умение загрузить картинку через **OpenCV**, перевести её в тензор, сделать предобработку (нормализацию, изменение размера) и прогнать через базовую CNN на **PyTorch**.
    - Понимание, что такое _Data Augmentation_ (аугментация) — как искусственно размножить датасет с помощью поворотов и изменения яркости снимков.

Тема 2: Детекция объектов (Object Detection) и семейство моделей YOLO

В робототехнике и видеоаналитике ритейла редко нужно просто сказать, «что на картинке» (классификация). Обычно нужно найти конкретный объект (коробку, человека, товар) и получить его **координаты (Bounding Box)**.

- **Что нужно знать:**
    - Разницу между Two-Stage детекторами (Faster R-CNN — точные, но медленные) и One-Stage детекторами (**YOLO, SSD** — супербыстрые, работают в реальном времени).
    - Как устроена метрика **IoU (Intersection over Union)** — пересечение предсказанной рамки с реальной.
    - Как работает алгоритм **NMS (Non-Maximum Suppression)**, который убирает дублирующиеся рамки вокруг одного и того же объекта.
- **Практический навык:**
    - Развернуть **YOLOv8 / YOLOv10**. Написать скрипт, который обрабатывает видеопоток с веб-камеры в реальном времени и рисует рамки вокруг объектов.

---

Тема 3: Transfer Learning (Тонкая настройка моделей) и Оптимизация инференса

В реальном бизнесе никто не обучает нейросети с нуля — на это нужны миллионы долларов и миллионы картинок. Инженеры берут готовые (pretrained) модели и дообучают их.

- **Что нужно знать:**
    - **Fine-tuning и Transfer Learning:** Как «заморозить» первые слои нейросети (которые умеют распознавать общие формы, линии и текстуры) и переобучить только финальные слои (голову/head) под специфические задачи (например, распознавание конкретных пачек молока ВкусВилл).
    - **Метрики качества:** Глубокое понимание **mAP (Mean Average Precision)** — главной метрики в детекции объектов.
    - **Экспорт и квантование:** Что такое **ONNX** и **TensorRT**. Как перевести веса модели из FP32 (числа с плавающей точкой) в INT8 (целые числа), чтобы модель весила в 4 раза меньше и летала на слабом железе, почти не теряя в точности.
