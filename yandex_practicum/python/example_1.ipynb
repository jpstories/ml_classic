## Строки  

**str: immutable** - любое форматирование создает новый обьект в памяти, объем зависит от длины и кодировки строки

```python
# 4 метода форматирования которые я знаю
print(str(123).rjust(9, "0")) # 2.30s [52, 58]
print("%09d" % 123)           # 2.52s [0, 58]
print(f"{123:0>9}")           # 3.41s [0, 58]
print("{:0>9}".format(123))   # 4.42s [~8, 58]
```

Как это работает по шагам:  
1. Python берет число 123. В нем 3 знака.  
2. Программа видит общую длину 9 и вычисляет, сколько символов не хватает: 9 - 3 = 6.  
3. Символ > указывает добавить эти 6 недостающих знаков слева от числа.  
4. Символ 0 указывает, что добавлять нужно именно нули, а не пробелы.

000000123 (ASCII, 9 символов) 58 байт

## Циклы

`while` - удобно использовать тогда, когда заранее не известно, сколько раз нужно повторять действие, — мы просто проверяем условие на каждой итерации.  

`:=` - моржовый оператор

```python
# До python 3.8
name = input("Введите имя: ")
while name != "СТОП":
    print(f"Привет, {name}!")
    name = input("Введите имя: ")
print("Программа завершена.")
# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
while (name := input("Введите имя: ") != "СТОП"):
    print(f"Привет, {name}!")
print("Программа завершена.")

# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
numbers = [9, 16, 25, 36, 49]
# math.sqrt(x) вызывается ОДИН раз для каждого элемента, а не два
roots = [root for x in numbers if (root := math.sqrt(x)) > 5]
# Результат: [6.0, 7.0]

# ====================================================================================================================================================================================
# break continue !=
flag = False
for i in range(26):
    for j in range(26):
        # if i == j:
        #     continue
        if i != j:
            print(f"{chr(ord('a') + i)}{chr(ord('a') + j)}")
        if i == 24:
            flag = True
            break
    if flag:
        break
# ====================================================================================================================================================================================
# else
while input("Введите строку (СТОП для остановки): ") != "СТОП":
    pass
else:
    print("Цикл завершён")
```

```python
# Наибольший
n = int(input())

# Считываем первое число и берем его за стартовый НОД
nod = int(input())

# Цикл делает на одну итерацию меньше, так как первое число мы уже считали
for i in range(n - 1):
    m = int(input())
    
    # Делаем копии, чтобы не сломать переменные nod и m
    x = nod
    y = m
    
    # Алгоритм Евклида
    while x > 0 and y > 0:
        if x > y:
            x %= y
        else:
            y %= x
            
    # Записываем новый найденный НОД для следующего шага
    nod = x + y

print(nod)
```

## Слоенный квадрат

```python
n = int(input())  # Размер матрицы

# 1. Вычисляем максимальное число, которое будет в самом центре
max_val = (n + 1) // 2

# 2. Находим ширину этого числа для правильного выравнивания
width = len(str(max_val))

for i in range(n):
    for j in range(n):
        # Находим расстояние до каждого из 4-х краёв
        dist_top = i
        dist_bottom = n - 1 - i
        dist_left = j
        dist_right = n - 1 - j
        
        # Вычисляем число для текущей ячейки
        layer = min(dist_top, dist_bottom, dist_left, dist_right)
        number = layer + 1
        
        # Если это последний столбец в строке, выводим без пробела на конце
        if j == n - 1:
            print(f"{number:>{width}}")
        else:
            print(f"{number:>{width}}", end=" ")
```

## Списки

Изменяемые коллекции - элементы можно изменить, удалять, добавлять.

`list` - это упорядоченная коллекция, в которой можно хранить значения любых типов: числа, строки, логические значения и даже другие списки.

```python
mixed_list = [10, 20, 30, 40, 50, "text"]

# del
del mixed_list[-1]  # [10, 20, 30, 40, 50]
del mixed_list[::2] # [20, 40]

# x in s
20 in mixed_list # True

# x not in s
20 not in mixed_list # False

# s + t
mixed_list + [60, 80] # [20, 40, 60, 80]

# s * n (n * s)
mixed_list * 2 # [20, 40, 60, 80, 20, 40, 60, 80]

# len(s), min(s), max(s)
len(mixed_list) # 2
min(mixed_list) # 20
max(mixed_list) # 40

# s.index_num(x)
mixed_list.index(20) # 0

# s.count(x)
mixed_list.count(20) # 1

# s.append(x)
mixed_list.append(60) # [20, 40, 60]

# s.clear()
mixed_list.clear() # []

# s.copy()
mixed_list.copy() # [20, 40, 60]

# s.extend(t) или s += t
mixed_list.extend([80, 100]) # [20, 40, 60, 80, 100]

# s.insert(i, x)
mixed_list.insert(1, 30) # [20, 30, 40, 60, 80, 100]

# s.pop(i), default=-1
x = mixed_list.pop() # 100 | [20, 30, 40, 60, 80]

# s.remove(x)
mixed_list.remove(30) # [20, 40, 60, 80]

# срез [::-1] - создает новый обьект
# s.reverse - меняет исходный обьект
mixed_list[::-1]
mixed_list.reverse() # [80, 60, 40, 20]

# s.sort()
mixed_list.sort()             # [20, 40, 60, 80]
mixed_list.sort(reverse=True) # [80, 60, 40, 20]

# sorted(s)
new_s = sorted(mixed_list, reverse=True) # [80, 60, 40, 20]
```

## Кортежи

Неизменяемые коллекции

`(tuple)` — это упорядоченная коллекция, как и список, но неизменяемая. После создания в кортеже нельзя изменить, удалить или добавить элементы. Это делает его удобным, когда нужно защитить данные от случайных изменений.

>Методы: len(), count(), index(), [ : : ], in

Кортежи удобно использовать, когда нужно временно объединить значения. Например, для обмена значениями двух переменных

```python
numbers = (1, 2, 3, 4, 5)

one_number = (1, ) # <class 'tuple'>
# one_number = (1)   # <class 'int'>

a = 1
b = 2
(a, b) = (b, a) # a = 2, b = 1
# можно опустить круглые скобки и записать так a, b = b, a
```
