## Строки  

**str: immutable** - любое форматирование создает новый обьект в памяти, объем зависит от длины и кодировки строки

```python
# 4 метода форматирования которые я знаю
print(str(123).rjust(9, "0")) # 2.30s [52, 58]
print("%09d" % 123)           # 2.52s [0, 58]
print(f"{123:0>9}")           # 3.41s [0, 58]
print("{:0>9}".format(123))   # 4.42s [~8, 58]
```

Как это работает по шагам:  
1. Python берет число 123. В нем 3 знака.  
2. Программа видит общую длину 9 и вычисляет, сколько символов не хватает: 9 - 3 = 6.  
3. Символ > указывает добавить эти 6 недостающих знаков слева от числа.  
4. Символ 0 указывает, что добавлять нужно именно нули, а не пробелы.

000000123 (ASCII, 9 символов) 58 байт

## Циклы

`while` - удобно использовать тогда, когда заранее не известно, сколько раз нужно повторять действие, — мы просто проверяем условие на каждой итерации.  

`:=` - моржовый оператор

```python
# До python 3.8
name = input("Введите имя: ")
while name != "СТОП":
    print(f"Привет, {name}!")
    name = input("Введите имя: ")
print("Программа завершена.")
# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
while (name := input("Введите имя: ") != "СТОП"):
    print(f"Привет, {name}!")
print("Программа завершена.")

# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
numbers = [9, 16, 25, 36, 49]
# math.sqrt(x) вызывается ОДИН раз для каждого элемента, а не два
roots = [root for x in numbers if (root := math.sqrt(x)) > 5]
# Результат: [6.0, 7.0]

# ====================================================================================================================================================================================
# break continue !=
flag = False
for i in range(26):
    for j in range(26):
        # if i == j:
        #     continue
        if i != j:
            print(f"{chr(ord('a') + i)}{chr(ord('a') + j)}")
        if i == 24:
            flag = True
            break
    if flag:
        break
# ====================================================================================================================================================================================
# else
while input("Введите строку (СТОП для остановки): ") != "СТОП":
    pass
else:
    print("Цикл завершён")
```

`for` - используется, когда количество итераций заранее известно. Он выполняет код для каждого элемента заданной последовательности.  

`range(start, stop, step)` — это не функция, а встроенный тип данных (Built-in Type), который создаёт последовательность целых чисел. (от start до stop - 1).

## НОД

```python
# Наименьший
n = int(input())
m = int(input())

while n != 0 and m != 0:
    if n > m:
        n = n % m
    else:
        m = m % n

nod = n + m
print(nod)
```

```python
# Наибольший
n = int(input())

# Считываем первое число и берем его за стартовый НОД
nod = int(input())

# Цикл делает на одну итерацию меньше, так как первое число мы уже считали
for i in range(n - 1):
    m = int(input())
    
    # Делаем копии, чтобы не сломать переменные nod и m
    x = nod
    y = m
    
    # Алгоритм Евклида
    while x > 0 and y > 0:
        if x > y:
            x %= y
        else:
            y %= x
            
    # Записываем новый найденный НОД для следующего шага
    nod = x + y

print(nod)
```

## Проверка Хэша в блокчейне

```python
n = int(input())

hn = 0
first = -1

for i in range(n):
    bn = int(input())
    h = bn % 256
    rn = (bn // 256) % 256
    mn = bn // (256 ** 2)
    h_new = 37 * (mn + rn + hn) % 256

    if (h != h_new or h >= 100) and first == -1:
        first = i
    
    hn = h
    
print(first)
```

## Вывод красивого прямоугольника цифрами строчно

```python
n = int(input())
m = int(input())

max_number = n * m
width = 0
while max_number:
    width += 1
    max_number //= 10

for row in range(n):
    for col in range(m):
        number = row * m + col + 1
        
        if col == m - 1:
            print(f"{number:>{width}}")
        else:
            print(f"{number:>{width}}", end=" ")
```

## Змейкой

```python
n = int(input())
m = int(input())

max_number = n * m
width = 0
while max_number:
    width += 1
    max_number //= 10

for row in range(n):
    for col in range(m):
        if row % 2 == 0:
            number = row * m + col + 1
        else:
            number = row * m + (m - 1 - col) + 1
        
        if col == m - 1:
            print(f"{number:>{width}}")
        else:
            print(f"{number:>{width}}", end=" ")
```

## Красивая таблица умножения

```python
n = int(input())
width = int(input())

line_length = n * width + 1 * (n - 1)

for i in range(1, n + 1):
    for j in range(1, n + 1):
        print(f"{i * j:^{width}}", end="")
        
        if j < n:
            print("|", end="")
            
    print("")
    
    if i < n:
        print("-" * line_length)
```

## Палиндром 

```python
n = int(input())
count = 0

for i in range(n):
    m = int(input())
    orig = m
    reversed_num = 0
    
    if m // 10 == 0:
        count += 1
    else:
        while m > 0:
            digit = m % 10
            reversed_num = reversed_num * 10 + digit
            m = m // 10

        if orig == reversed_num:
            count += 1
            
print(count)
```

## Ёлочка 2.0

```python
max_number = int(input())

row = 1
col = 1
number = 1
row_value = ''

while number <= max_number:
    row_value = ''
    while col <= row and number <= max_number:
        row_value += str(number) + (' ' if col < row and number < max_number else '')
        col += 1
        number += 1
    row += 1
    col = 1

total_width = len(row_value)

row = 1
col = 1
number = 1

while number <= max_number:
    row_value = ''
    while col <= row and number <= max_number:
        row_value += str(number) + (' ' if col < row and number < max_number else '')
        col += 1
        number += 1
    
    print(f"{row_value:^{total_width}}")
    
    row += 1
    col = 1
```

## Слоенный квадрат

```python
n = int(input())  # Размер матрицы

# 1. Вычисляем максимальное число, которое будет в самом центре
max_val = (n + 1) // 2

# 2. Находим ширину этого числа для правильного выравнивания
width = len(str(max_val))

for i in range(n):
    for j in range(n):
        # Находим расстояние до каждого из 4-х краёв
        dist_top = i
        dist_bottom = n - 1 - i
        dist_left = j
        dist_right = n - 1 - j
        
        # Вычисляем число для текущей ячейки
        layer = min(dist_top, dist_bottom, dist_left, dist_right)
        number = layer + 1
        
        # Если это последний столбец в строке, выводим без пробела на конце
        if j == n - 1:
            print(f"{number:>{width}}")
        else:
            print(f"{number:>{width}}", end=" ")
```
