## Строки  

**str: immutable** - любое форматирование создает новый обьект в памяти, объем зависит от длины и кодировки строки

```python
# 4 метода форматирования которые я знаю
print(str(123).rjust(9, "0")) # 2.30s [52, 58]
print("%09d" % 123)           # 2.52s [0, 58]
print(f"{123:0>9}")           # 3.41s [0, 58]
print("{:0>9}".format(123))   # 4.42s [~8, 58]
```

Как это работает по шагам:  
1. Python берет число 123. В нем 3 знака.  
2. Программа видит общую длину 9 и вычисляет, сколько символов не хватает: 9 - 3 = 6.  
3. Символ > указывает добавить эти 6 недостающих знаков слева от числа.  
4. Символ 0 указывает, что добавлять нужно именно нули, а не пробелы.

`while` - удобно использовать тогда, когда заранее не известно, сколько раз нужно повторять действие, — мы просто проверяем условие на каждой итерации.  

`:=` - моржовый оператор

```python
# До python 3.8
name = input("Введите имя: ")
while name != "СТОП":
    print(f"Привет, {name}!")
    name = input("Введите имя: ")
print("Программа завершена.")
# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
while (name := input("Введите имя: ") != "СТОП"):
    print(f"Привет, {name}!")
print("Программа завершена.")

# ====================================================================================================================================================================================

# Моржовый оператор :=, создание + проверка
numbers = [9, 16, 25, 36, 49]
# math.sqrt(x) вызывается ОДИН раз для каждого элемента, а не два
roots = [root for x in numbers if (root := math.sqrt(x)) > 5]
# Результат: [6.0, 7.0]

# ====================================================================================================================================================================================
# break continue !=
flag = False
for i in range(26):
    for j in range(26):
        # if i == j:
        #     continue
        if i != j:
            print(f"{chr(ord('a') + i)}{chr(ord('a') + j)}")
        if i == 24:
            flag = True
            break
    if flag:
        break
# ====================================================================================================================================================================================
# else
while input("Введите строку (СТОП для остановки): ") != "СТОП":
    pass
else:
    print("Цикл завершён")
```

```python
# Наибольший
n = int(input())

# Считываем первое число и берем его за стартовый НОД
nod = int(input())

# Цикл делает на одну итерацию меньше, так как первое число мы уже считали
for i in range(n - 1):
    m = int(input())
    
    # Делаем копии, чтобы не сломать переменные nod и m
    x = nod
    y = m
    
    # Алгоритм Евклида
    while x > 0 and y > 0:
        if x > y:
            x %= y
        else:
            y %= x
            
    # Записываем новый найденный НОД для следующего шага
    nod = x + y

print(nod)
```

## Списки

Изменяемые коллекции - элементы можно изменить, удалять, добавлять.

*Занимает больше памяти*, так как резервирует место под новые элементы.

`list` - это упорядоченная коллекция, в которой можно хранить значения любых типов: числа, строки, логические значения и даже другие списки.

```python
mixed_list = [10, 20, 30, 40, 50, "text"]

# del
del mixed_list[-1]  # [10, 20, 30, 40, 50]
del mixed_list[::2] # [20, 40]

# x in s
20 in mixed_list # True

# x not in s
20 not in mixed_list # False

# s + t
mixed_list + [60, 80] # [20, 40, 60, 80]

# s * n (n * s)
mixed_list * 2 # [20, 40, 60, 80, 20, 40, 60, 80]

# len(s), min(s), max(s)
len(mixed_list) # 2
min(mixed_list) # 20
max(mixed_list) # 40

# s.index_num(x)
mixed_list.index(20) # 0

# s.count(x)
mixed_list.count(20) # 1

# s.append(x)
mixed_list.append(60) # [20, 40, 60]

# s.clear()
mixed_list.clear() # []

# s.copy()
mixed_list.copy() # [20, 40, 60]

# s.extend(t) или s += t
mixed_list.extend([80, 100]) # [20, 40, 60, 80, 100]

# s.insert(i, x)
mixed_list.insert(1, 30) # [20, 30, 40, 60, 80, 100]

# s.pop(i), default=-1
x = mixed_list.pop() # 100 | [20, 30, 40, 60, 80]

# s.remove(x)
mixed_list.remove(30) # [20, 40, 60, 80]

# срез [::-1] - создает новый обьект
# s.reverse - меняет исходный обьект
mixed_list[::-1]
mixed_list.reverse() # [80, 60, 40, 20]

# s.sort()
mixed_list.sort()             # [20, 40, 60, 80]
mixed_list.sort(reverse=True) # [80, 60, 40, 20]

# sorted(s)
new_s = sorted(mixed_list, reverse=True) # [80, 60, 40, 20]
```

## Кортежи

Неизменяемые коллекции

`(tuple)` — это упорядоченная коллекция, как и список, но неизменяемая. После создания в кортеже нельзя изменить, удалить или добавить элементы. Это делает его удобным, когда нужно защитить данные от случайных изменений.

*Занимает меньше памяти* и работает быстрее.

>Методы: len(), count(), index(), [ : : ], in

Кортежи удобно использовать, когда нужно временно объединить значения.  
a, b = b, a

```python
numbers = (1, 2, 3, 4, 5)

one_number = (1, ) # <class 'tuple'>
# one_number = (1)   # <class 'int'>

a = 1
b = 2
(a, b) = (b, a) # a = 2, b = 1
# можно опустить круглые скобки и записать так a, b = b, a
```

```python
s = input().split(" ")
stack = []

for char in s:
    if char in ('+', '-', '*'):
        # Достаем два последних операнда из стека
        # Важно: первыми извлекается правый операнд (b), затем левый (a)
        b = stack.pop()
        a = stack.pop()
        
        if char == '+':
            stack.append(a + b)
        elif char == '-':
            stack.append(a - b)
        elif char == '*':
            stack.append(a * b)
    else:
        # Если это число, преобразуем в int и кладем в стек
        stack.append(int(char))

# Единственное оставшееся число в стеке — результат вычисления
print(stack[0])
```

```python
s = input().split(" ")

stack = []

for char in s:
    if char in ('+', '-', '*', '/'):
        b = stack.pop()
        a = stack.pop()
        
        if char == '+':
            stack.append(a + b)
        elif char == '-':
            stack.append(a - b)
        elif char == '*':
            stack.append(a * b)
        elif char == '/':
            stack.append(a // b)
            
    elif char in ('~', '!', '#'):
        c = stack.pop()

        if char == '~':
            stack.append(-c)
        elif char == '!':
            factorial = 1
            for i in range(1, c + 1):
                factorial *= i
            stack.append(factorial)
        elif char == '#':
            stack.append(c)
            stack.append(c)

    elif char == '@':
        d, f, g = stack.pop(), stack.pop(), stack.pop()

        stack.append(f)
        stack.append(d)
        stack.append(g)
    else:
        stack.append(int(char))

print(stack[0])
```

```python
translit_dict = {
    'А': 'A', 'Б': 'B', 'В': 'V', 'Г': 'G', 'Д': 'D', 'Е': 'E', 'Ё': 'E', 
    'Ж': 'ZH', 'З': 'Z', 'И': 'I', 'Й': 'I', 'К': 'K', 'Л': 'L', 'М': 'M', 
    'Н': 'N', 'О': 'O', 'П': 'P', 'Р': 'R', 'С': 'S', 'Т': 'T', 'У': 'U', 
    'Ф': 'F', 'Х': 'KH', 'Ц': 'TC', 'Ч': 'CH', 'Ш': 'SH', 'Щ': 'SHCH', 
    'Ы': 'Y', 'Э': 'E', 'Ю': 'IU', 'Я': 'IA',
    'Ь': '', 'Ъ': '', 'ь': '', 'ъ': ''
}

ru_text = input()

translit_text = []

for char in ru_text:
    u_char = char.upper()

    if u_char in translit_dict:
        replacement = translit_dict[u_char]
        if char.isupper():
            translit_char = replacement.capitalize()
        else:
            translit_char = replacement.lower()
        translit_text.append(translit_char)
    else:
        translit_text.append(char)

print("".join(translit_text))
```

## Карта сокровищ

```python
n = int(input())

# Обычный словарь для подсчета групп
counts = {}
max_points = 0

for i in range(n):
    # Считываем строку и сразу преобразуем в два числа
    x, y = input().split()
    x = int(x)
    y = int(y)
    
    # Отбрасываем только последнюю цифру у каждой координаты
    base_x = x // 10
    base_y = y // 10
    
    # Объединяем базовые координаты в кортеж (это будет ключ для словаря)
    key = (base_x, base_y)
    
    # Если такой ключ уже есть в словаре, увеличиваем счетчик
    if key in counts:
        counts[key] += 1
    # Если ключа еще нет, создаем его со значением 1 (первая точка)
    else:
        counts[key] = 1
        
    # Сразу обновляем максимальное количество точек
    if counts[key] > max_points:
        max_points = counts[key]

# Выводим размер самой большой группы
print(max_points)
```

```python
n = int(input())

# Словарь для подсчета количества владельцев каждой игрушки
toy_counts = {}

for _ in range(n):
    line = input()
    # Отрезаем имя, оставляем только строку с игрушками
    _, toys_str = line.split(":", 1)
    
    # Собираем уникальные игрушки для ТЕКУЩЕГО ребенка в множество.
    # set нужен на случай, если кто-то записал: "Вася: мяч, кукла, мяч"
    child_toys = {toy.strip() for toy in toys_str.split(",")}
    
    # Для каждой игрушки ребенка увеличиваем счетчик её владельцев
    for toy in child_toys:
        if toy in toy_counts:
            toy_counts[toy] += 1
        else:
            toy_counts[toy] = 1

# Собираем список игрушек, счетчик которых равен ровно 1
unique_toys = []
for toy, count in toy_counts.items():
    if count == 1:
        unique_toys.append(toy)

# Сортируем по алфавиту и выводим
for toy in sorted(unique_toys):
    print(toy)
```

```python
# Считываем строку, разбиваем по ";", очищаем от пробелов, 
# переводим в числа, убираем дубликаты через set и сортируем (sorted)
input_data = input().split(";")
numbers = sorted(set(int(n.strip()) for n in input_data))

# Проходимся по каждому числу
for num in numbers:
    coprimes = []
    
    # Сравниваем его со всеми остальными числами
    for other in numbers:
        if num == other:
            continue
            
        # Реализуем алгоритм Евклида для поиска НОД прямо здесь
        a = num
        b = other
        while b != 0:
            a, b = b, a % b
            
        # После окончания цикла while в переменной 'a' хранится НОД
        # Если НОД равен 1, значит числа взаимно простые
        if a == 1:
            coprimes.append(other)
            
    # Если нашлось хотя бы одно взаимно простое число, выводим результат
    if coprimes:
        coprimes_str = ", ".join(str(c) for c in coprimes)
        print(f"{num} - {coprimes_str}")
```

## Генераторы

В Python генераторы позволяют создавать последовательности значений «на лету» — без хранения всех элементов в памяти. Они похожи на списочные выражения, но вместо квадратных скобок используют круглые:

```python
numbers = (int(input()) for i in range(5))
print(numbers) # <generator object <genexpr> at 0x000001F2D1DFAF80>
```

`Ленивые вычисления`: Генератор не создает миллион чисел сразу. Он лишь хранит текущее состояние, инструкции для создания следующего числа и ссылку на объект range.  

`Фиксированный размер`: Размер генератора в байтах всегда одинаков и не зависит от того, сколько элементов он должен обработать — 10 ** 6 или 10 ** 12.  

>Начиная с Python 3.11 и новее разработчики CPython добавили в объект генератора (PyGenObject на уровне C) новые внутренние поля для ускорения работы кода. Из-за этого его базовый размер вырос до c 112 байт до 200 байт. (в 32-битных системах: 56-64 байт)

```python
import sys

numbers_iter = (i for i in range(10 ** 6))
sys.getsizeof(numbers_iter) # 200 байт

numbers_list = list(range(10 ** 6))
sys.getsizeof(numbers_list) # 8000056 байт! (7.6 мб)
```

Генератор может быть быстрее, чем список, при условии обработки большого количества элементов.

```python
from timeit import timeit

print(round(timeit("s = '; '.join(str(x) for x in range(10 ** 7))", number=10), 3))
print(round(timeit("s = '; '.join([str(x) for x in list(range(10 ** 7))])", number=10), 3))
```

## Изменяемые и неизменяемые типы в Python

В Python абсолютно всё является объектом, `изменяемым` или `неизменяемым`.

**неизменяемые** - значения меняются, ссылка тоже  
**изменяемые** - значения не меняются, ссылка тоже не меняется

В Python переменные хранят не данные, а ярлыки (`ссылки`) на объекты в области памяти.  

У каждого объекта есть `идентификатор` — уникальный номер, который можно узнать с помощью функции `id()`

```python
# Числа в Python неизменяемы (immutable)

# создает в памяти объект-int
# Ярлык x начинает указывать на этот объект
x = 5
id(x) # 640

# Python не может зайти по старому адресу и переписать 5 на 10
# Вместо этого он создает в памяти новый объект со значением 10 и перенаправляет ярлык x на него.
x = 10
id(x) # 800
```

>CPython ради экономии памяти заранее создает (кэширует) объекты для чисел от -5 до 256. Поэтому любые переменные, равные 5, всегда будут ссылаться на один и тот же объект в области памяти

```python
a = 5
b = 5
id(a) == id(b) # True
id(a) is id(b) # False

x = 1
y = x
id(x) # 140716947657512
id(y) # 140716947657512
x is y # True
x == y # True
```

```python
# Значения одинаковые, но x и y — разные списки в памяти, у них разные идентификаторы.
x = [el ** 2 for el in range(5)]
y = [el ** 2 for el in range(5)]
print(x == y) # True
print(x is y) # False
```

### Изменяемые и неизменяемые типы

- Неизменяемые типы: `int`, `float`, `str`, `tuple`, `frozenset`. (создаётся новый объект с новым идентификатором)  

- Изменяемые типы: `list`, `set`, `dict`. (можно изменить значение, не меняя идентификатор)

```python
numbers = [1, 2, 3]     # id 1632
numbers += [4]          # id 1632

numbers = numbers + [4] # id 8571!
```

```python
# Присваивание ссылки объекта
x = [1, 2, 3] # [0, 2, 3]
y = x         # [0, 2, 3]
x[0] = 0


# Срез создает новый обьект!
x = [1, 2, 3] # [0, 2, 3]
y = x[:]      # [1, 2, 3]
x[0] = 0
```
