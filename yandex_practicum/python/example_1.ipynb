## Строки  

**str: immutable** - любое форматирование создает новый обьект в памяти, объем зависит от длины и кодировки строки

```python
# 4 метода форматирования которые я знаю
print(str(123).rjust(9, "0")) # 2.30s [52, 58]
print("%09d" % 123)           # 2.52s [0, 58]
print(f"{123:0>9}")           # 3.41s [0, 58]
print("{:0>9}".format(123))   # 4.42s [~8, 58]
```

Как это работает по шагам:  
1. Python берет число 123. В нем 3 знака.  
2. Программа видит общую длину 9 и вычисляет, сколько символов не хватает: 9 - 3 = 6.  
3. Символ > указывает добавить эти 6 недостающих знаков слева от числа.  
4. Символ 0 указывает, что добавлять нужно именно нули, а не пробелы.

000000123 (ASCII, 9 символов) 58 байт

## Числа

```python
n = 10 # целое число
pi = 3.14 # вещественное число

# В CPython числа в диапазоне от -5 до 256 заранее созданы и постоянно живут в памяти

print(int("  42  ")) # 42 - указатель уже существует, возвращает указатель 
print(int("A", 16))  # 10
print(int("ff", 16)) # 255

print(int("1001", 2)) # 1 * 2^0 + 0 * 2^1 + 0 * 2^2 + 1 * 2^3 = 9
print(bin(9)) # bin 2, oct 8, hex 16


print(float("  -3.14 ")) # Вывод: -3.14
print(float("4.2e-2")) # 4.2 * 10^-2 -> Вывод: 0.042
# Для поиска минимума, Графы, Алгоритмы ветвей и границ (Minimax в шахматах - гарант. победа)
print(float("inf"))   # Бесконечность -> Вывод: inf 
print(float("-inf"))  # Минус бесконечность -> Вывод: -inf -> Для поиска максимума
print(float("nan"))   # Не-число -> Вывод: nan


print(4 / 2)  # float 2.0
print(4 // 2) # int   2

print(5.0 % 2)  # float 1
print(5 % 2)    # int   1

print(f"{2 ** 0.5:.2f}") # 1.41
print(1234 % 10)         # 4
print(1234 // 10 % 10)   # 3

print(1 ** 1 * 1 / 1 % 1 // 1 + 1 - 1) # Приоритет операций слева направо

```
