создаёт множество всех чисел, встречающихся в списке ровно один раз.

```python
numbers = [1, 2, 1, 3, 1, 2, 2, 4, 1, 2, 5, 1, 2]
{number for number in numbers if numbers.count(number) == 1}
```

находит максимальное произведение двух различных чисел

```python
numbers = {2, 4, 5, 7, -10, -8, 10, -9, -1}
max(x * y for x in numbers for y in numbers if x != y)
```

находит ключ с наименьшей суммой значений

```python
data = {'a': [1, 2, 3], 'b': [2, 3, 4, 5]}
# (6, 'a')

min((sum(v), k) for k, v in data.items())[1]
```

```python
data = {'a': [1, 2, 3], 'b': [5, 2, 5], 'c': [7, 15, 3]}
print({k for k, v in data.items() if len(set(v)) != len(v)})
```

частота употребления букв в тексте

```python
text = 'Мама мыла раму!'
print({char: text.lower().count(char) for char in set(text.lower()) if char.isalpha()})
```

RLE

```python
rle = [('a', 2), ('b', 3), ('c', 1)]
print("".join(char * count for char, count in rle))
```

```python
numbers = {15, 49, 36}
print({k: [num for num in range(1, k + 1) if k % num == 0] for k in numbers})
```

Простые числа

```python
numbers = {1, 2, 3, 4, 5}
{num for num in numbers if num > 1 and not [i for i in range(2, int(num ** 0.5) + 1) if num % i == 0]}
```
