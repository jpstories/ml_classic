```python
from itertools import product, count, cycle, repeat, accumulate, chain

# Без product
for a in range(10):
    for b in range(10):
        for c in range(10):
            for d in range(10):
                for e in range(10):
                    print(a, b, c, d, e)

# С product — глубина вложенности всего 1 уровень
for a, b, c, d, e in product(range(10), repeat=5):
    print(a, b, c, d, e)

# ================================================

# 2. count()
for value in count(0, 0.1):
    if value > 1:
        break

    print(round(value, 1))

# 3. cycle()
s = ""
for letter in cycle("ABC"):
    if len(s) == 10:
        break
    s += letter

print(s)

# 4. repeat()
print(list(repeat("ABC", 5)))

# 5. accumulate()
# Полезно, если нужно посчитать прогресс, баллы
for value in accumulate([1, 2, 3, 4, 5]):
    print(value)

# 6. chain() и chain.from_iterable()
list_a = [1, 2, 3]
tuple_b = (4, 5)
set_c = {6, 7}
combined1 = chain(list_a, tuple_b, set_c)
combined2 = chain.from_iterable([list_a, tuple_b, set_c])
print(list(combined1))
print(list(combined2))
```

#### Колода карт

```python
from itertools import product

suit_exclude = input().strip()

suits = ["пик", "треф", "бубен", "червей"]
weights = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "валет", "дама", "король", "туз"]

if suit_exclude in suits:
    suits.remove(suit_exclude)

for weight, suit in product(weights, suits):
    print(f"{weight} {suit}")
    
```

#### Список всех возможных партий между игроками

```python
from itertools import combinations

nums = int(input())
names = [input().strip() for _ in range(nums)]

for player_a, player_b in combinations(names, 2):
    print(f"{player_a} - {player_b}")
```

```python
from itertools import cycle, islice

m = int(input())
porridges = [input().strip() for _ in range(m)]
days = int(input())

inf_porridge = cycle(porridges)

porridge_in_day = islice(inf_porridge, days)

for porridge in porridge_in_day:
    print(porridge)
```

#### Таблица умножения

```python
from itertools import product

n = int(input())

for i, j in product(range(1, n + 1), repeat=2):
    if j == n:
        print(i * j)
    else:
        print(i * j, end=" ")
```

#### Разделить апельсин

```python
from itertools import product

n = int(input())

print("А Б В")

possible_slices = range(1, n - 1)

for a, b in product(possible_slices, repeat=2):
    v = n - a - b

    if v >= 1:
        print(f"{a} {b} {v}")
```

#### прямоугольник

```python
rows = int(input())
cols = int(input())

linear = range(1, rows * cols + 1)

width = len(str(rows * cols))

iters = [iter(linear)] * cols
result = list(zip(*iters))

for tpl in result:
    formatted_row = [f"{x:>{width}}" for x in tpl]
    print(" ".join(formatted_row))
```

```python
import itertools

required_suit_nom = input().strip()
forbidden_rank = input().strip()

suits_genitive = {
    "буби": "бубен",
    "пики": "пик",
    "трефы": "треф",
    "черви": "червей"
}

req_suit = suits_genitive[required_suit_nom]

ranks = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "валет", "дама", "король", "туз"]
suits = ["бубен", "пик", "треф", "червей"]

deck = []
for rank in ranks:
    if rank != forbidden_rank:
        for suit in suits:
            deck.append(f"{rank} {suit}")

valid_combinations = []

for combo in itertools.combinations(deck, 3):
    has_req_suit = False
    for card in combo:
        if card.endswith(req_suit):
            has_req_suit = True
            break
            
    if has_req_suit:
        sorted_combo = sorted(combo)
        combo_str = ", ".join(sorted_combo)
        valid_combinations.append(combo_str)

valid_combinations.sort()

for combo in valid_combinations[:10]:
    print(combo)
```

#### Файл — это именованная область данных на диске  
Данные в нём хранятся как последовательность байтов. Для корректной работы с символами нужно знать **кодировку**.

Функция `open()`

```python
open(file, mode='r', encoding=None)
```

Аргумент `mode` определяет режим доступа к файлу:

- `'r'` — чтение (по умолчанию).  
    Файл должен существовать, иначе будет ошибка.
- `'w'` — запись с очисткой содержимого.  
    Если файл существует, его содержимое удаляется. Если не существует — создаётся.
- `'a'` — добавление в конец файла.  
    Новый текст будет дописан после уже имеющегося. Если файл не существует — создаётся.
- `'r+'` — чтение и запись, без удаления содержимого.  
    Файл должен существовать. Запись идёт поверх существующего текста, начиная с начала файла.
- `'w+'` — чтение и запись, с удалением содержимого.  
    Файл очищается или создаётся заново. Подходит, если вы хотите сначала записать, а потом прочитать.
- `'a+'` — чтение и добавление.  
    Файл открывается в режиме добавления, но можно также его читать. При записи — курсор всегда в конце.

> Важно: в Windows по умолчанию используется кодировка cp1251, поэтому нужно всегда явно указывать кодировку UTF-8, если вы хотите, чтобы программа работала кросс-платформенно.

```python
file_in = open("input_1.txt", encoding="UTF-8")
for line in file_in:
    print(line.rstrip("\n"))
file_in.close()
```

#### Важно использовать менеджер контекста with при работе с файлами  
Обычно после работы с файлом его нужно **закрывать вручную**

```python
file_in.close()
```

Если этого не сделать, файл может остаться открытым:

- он будет заблокирован для других программ;
- данные могут не сохраниться сразу;
- при возникновении ошибки до завершения работы с файлом, он останется открытым.

> with гарантирует, что файл будет закрыт автоматически.

```python
with open("input_1.txt", encoding="UTF-8") as file_in:
    for line in file_in:
        print(line.rstrip("\n"))
```

Следует помнить о том, что устройство хранения данных является самым *медленным* в компьютере, так что частые операции с файлами будут снижать скорость работы вашей программы.

Поэтому, если позволяют ресурсы компьютера (объём оперативной памяти), файл лучше прочитать сразу целиком, работать с ним в оперативной памяти и записывать в файл полностью полученный результат.

```python
# Чтение файла целиком — списком строк
with open("input_1.txt", encoding="UTF-8") as file_in:
    lines = file_in.readlines()
print(lines)

# Вывод программы:
# ['Привет!\n', 'Это пример текстового файла.\n', 'А это -- последняя строка, которая должна закончиться переходом на новую.\n']


# Чтение первых символов файла
with open("input_1.txt", encoding="UTF-8") as file_in:
    symbols = file_in.read(10)
print([symbols])

# Вывод программы:
# ['Привет!\nЭт']
```

```python
with open("output_1.txt", "w", encoding="UTF-8") as file_out:
    n = file_out.write("Это первая строка\nА вот и вторая\nИ третья -- последняя\n")

print(n) # 55

#====================================================================================

lines = ["Это первая строка\n", "А вот и вторая\n", "И третья - последняя\n"]
with open("output_2.txt", "w", encoding="UTF-8") as file_out:
    file_out.writelines(lines)

# Содержимое выходного файла:
# Это первая строка
# А вот и вторая
# И третья - последняя

#====================================================================================

with open("output_3.txt", "w", encoding="UTF-8") as file_out:
    print("Вывод в файл с помощью функции print()", file=file_out)

# Содержимое выходного файла:
# Вывод в файл с помощью функции print()
```

### JSON  
Модуль json автоматически преобразует ключи словаря в строки
- load()
- dump()

```python
import json
from pprint import pprint

with open("data.json", encoding="UTF-8") as file_in:
    records = json.load(file_in)

records[1]["group_number"] = 5

with open("data.json", "w", encoding="UTF-8") as file_out:
    json.dump(records, file_out, ensure_ascii=False, indent=2)
```
