```python
from itertools import product, count, cycle, repeat

# 1. product()
print(product("ABC", repeat=2))

# 2. count()
for value in count(0, 0.1):
    if value > 1:
        break

    print(round(value, 1))

# 3. cycle()
s = ""
for letter in cycle("ABC"):
    if len(s) == 10:
        break
    s += letter

print(s)

# 4. repeat()
print(list(repeat("ABC", 5)))

# Вывод:
# ['ABC', 'ABC', 'ABC', 'ABC', 'ABC']
```
