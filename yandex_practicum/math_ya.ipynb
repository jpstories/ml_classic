```python
import math
import matplotlib.pyplot as plt
import numpy as np

# Генерируем данные с помощью numpy для плавных линий
x_trig = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
x_inv = np.linspace(-1, 1, 500)
x_atan = np.linspace(-5, 5, 500)

# Настройка графиков
fig, axs = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Тригонометрические и обратные функции в Python', fontsize=16, fontweight='bold')

# --- 1. Синус ---
axs[0, 0].plot(x_trig, np.sin(x_trig), color='blue', label='sin(x)')
axs[0, 0].set_title('math.sin(x)')
axs[0, 0].set_ylim(-1.5, 1.5)

# --- 2. Косинус ---
axs[0, 1].plot(x_trig, np.cos(x_trig), color='green', label='cos(x)')
axs[0, 1].set_title('math.cos(x)')
axs[0, 1].set_ylim(-1.5, 1.5)

# --- 3. Тангенс ---
# Маскируем разрывы асимптот, чтобы не было вертикальных линий
y_tan = np.tan(x_trig)
y_tan[abs(y_tan) > 10] = np.nan
axs[0, 2].plot(x_trig, y_tan, color='red', label='tan(x)')
axs[0, 2].set_title('math.tan(x)')
axs[0, 2].set_ylim(-5, 5)

# --- 4. Арксинус ---
axs[1, 0].plot(x_inv, np.arcsin(x_inv), color='purple', label='asin(x)')
axs[1, 0].set_title('math.asin(x)')
axs[1, 0].set_ylabel('Ограничено от -π/2 до π/2')

# --- 5. Арккосинус ---
axs[1, 1].plot(x_inv, np.arccos(x_inv), color='orange', label='acos(x)')
axs[1, 1].set_title('math.acos(x)')
axs[1, 1].set_ylabel('Ограничено от 0 до π')

# --- 6. Арктангенс ---
axs[1, 2].plot(x_atan, np.arctan(x_atan), color='brown', label='atan(x)')
axs[1, 2].set_title('math.atan(x)')

# Общая стилизация для всех осей
for ax in axs.flat:
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import math

x = float(input("Введите x: "))

# Разбор формулы по частям:
# 1. math.log(аргумент, основание) -> math.log(x**3 / 16, 32)
# 2. x в степени cos(pi * x / 2) * e -> x ** math.cos(math.pi * x / 2) * math.e
# 3. Квадрат синуса (sin(x / pi))**2 -> math.sin(x / math.pi) ** 2

y = (math.log(x ** 3 / 16, 32) 
     + (x ** math.cos(math.pi * x / 2)) * math.e 
     - math.sin(x / math.pi) ** 2)

print(y)
```

```python
import math

n, m = map(int, input().split())

# Если мест больше или столько же, сколько людей, то все варианты успешны
if m >= n:
    total = 1
    fav = 1
else:
    # Всего вариантов собрать группу из M человек среди N зрителей
    total = math.comb(n, m)
    # Способы добрать группу из M-1 человек среди оставшихся N-1 (Вася уже внутри)
    fav = math.comb(n - 1, m - 1)
    
print(fav, total)
```

```python
import math

# 1. Считываем координаты первой точки в декартовой системе (x, y)
x1, y1 = map(float, input().strip().split())

# 2. Считываем координаты второй точки в полярной системе (p = ро, f = фи)
p, f = map(float, input().strip().split())

# 3. Переводим полярные координаты второй точки в декартовы
x2_d = p * math.cos(f)
y2_d = p * math.sin(f)

# 4. Считаем расстояние между двумя точками в декартовой системе
path = math.dist((x1, y1), (x2_d, y2_d))

print(f"{path:.2f}" if path % 1 == 0 or len(str(path).split('.')[1]) == 1 else path)
```
