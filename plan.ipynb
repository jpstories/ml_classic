### Фаза 1. Качественный сдвиг в инженерию (1–2 недели)

Недостаточно обучать модели в Jupyter Notebook. Чтобы быть востребованным, нужно уметь превращать модель в работающий сервис.

- **Что изучить:**
    
    - **FastAPI:** Оберни обученный CatBoost или Random Forest в REST API сервис.
        
    - **Docker:** Упакуй модель и FastAPI-сервер в изолированный контейнер.
        
    - **Git & Модульный код:** Перенеси код из `.ipynb` ноутбуков в чистые `.py` модули (`train.py`, `predict.py`, `config.py`).
        
- **Практический результат:** Модель, которую можно запустить одной командой `docker run` и отправить на неё запрос через `curl` или Postman.
    

### Фаза 2. Переход в Deep Learning через PyTorch (3–4 недели)

Переход от `scikit-learn` к `PyTorch`. Здесь ты поймешь, как работают нейронные сети под капотом.

- **Что изучить:**
    
    - **Основы PyTorch:** Тензоры, автоматическое дифференцирование (`autograd`), написание своего `Dataset` и `DataLoader`, цикл обучения (`training loop`).
        
    - **Архитектуры:**
        
        - Полносвязные сети (MLP) — для знакомства с весами и функцией потерь.
            
        - Архитектура **Transformer (Self-Attention)** — абсолютный фундамент всех современных текстовых, голосовых и визуальных моделей (GPT, LLaMA, Whisper).
            
- **Практический результат:** Написанная с нуля на PyTorch простая сеть, обученная на текстах или изображениях.
    

### Фаза 3. Прикладной Generative AI & RAG (3–4 недели)

Самая востребованная ниша на рынке прямо сейчас для роли **AI Engineer**.

- **Что изучить:**
    
    - **Экосистема Hugging Face:** Загрузка и запуск готовых моделей через `transformers` и `pipeline`.
        
    - **RAG (Retrieval-Augmented Generation):** Интеграция базы знаний с LLM.
        
        - Векторные БД (**Qdrant** или **pgvector** в PostgreSQL).
            
        - Эмбеддинги (векторизация текста).
            
    - **Оркестрация агентов:** Изучи **LangGraph** или **LlamaIndex** для создания агентов, умеющих вызывать внешние функции (Tool Calling / Function Calling).
        
- **Практический результат:** Чат-бот, который ответит на вопросы по твоим личным PDF-документам с указанием источников.
    

### Фаза 4. Дообучение и оптимизация (Fine-Tuning & Inference) (3–4 недели)

Уровень, который отличает джуниора от мидла.

- **Что изучить:**
    
    - **Эффективное дообучение (PEFT / QLoRA):** Как дообучить модель на 8B параметров на одной обычной видеокарте, используя LoRA-адаптеры.
        
    - **Оптимизация инференса:** Использование движка **vLLM** или **Ollama** для ускорения генерации токенов в разы.
        
    - **Автоматическое оценивание (Evals):** Использование библиотек вроде `Ragas` или паттерна _LLM-as-a-Judge_ для проверки качества ответов модели.
        

## 🛠 Пример идеального «Pet-проекта» на стыке технологий

Чтобы выделиться среди кандидатов, сделай гибридный проект, объединяющий классический ML и LLM:

> **Система умной оценки скоринга клиентов:**
> 
> 1. **Классический ML (CatBoost):** Считает вероятность дефолта клиента по табличным данным (доход, возраст, кредитная история).
>     
> 2. **LLM + RAG (LangGraph + Qdrant):** Берет выписку из банка и текстовые комментарии менеджера, ищет внутренние регламенты банка и генерирует понятное текстовое резюме: _«Почему клиенту одобрен/отклонен кредит»_.
>     
> 3. **Инфраструктура:** Все упаковано в **Docker**, обернуто в **FastAPI** с веб-интерфейсом на Streamlit.
>     

Этот проект покажет работодателю, что ты умеешь работать и со строгими табличными данными, и с современными языковыми моделями, и с продакшен-инфраструктурой.
