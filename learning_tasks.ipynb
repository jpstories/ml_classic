1. **pgvector** Напишите на Python (FastAPI) сервис, который берет медицинский справочник, превращает его в векторы, сохраняет в Postgres и делает поиск «по смыслу» -> локальный RAG.

2. Концепция **Function Calling**. Полностью сфокусируйтесь на фреймворке **LangGraph**. Забудьте про линейный LangChain. Вам нужно научиться строить циклические графы: _Агент написал код ➔ Узел тестирования проверил ➔ Если ошибка, вернул агенту на доработку_.

3. Напишите Мультиагентная система на LangGraph для анализа медицинских или финансовых отчетов. Один агент парсит PDF, второй ищет аномалии через векторную базу (Postgres + pgvector), третий пишет отчет на основе бустинга (объедините ваш старый опыт ML и новые LLM). Сделайте красивый фронтенд.
Обязательно упакуйте всё в **Docker**, настройте логирование и напишите для системы тестовый стенд (**Harness**), который проверяет точность работы агентов.  

4. Разберитесь, как работают облачные движки (Groq, vLLM) и локальная Ollama. Научитесь методам сжатия контекста (Conversation Summary), о которых мы говорили. Ваша система должна уметь работать как на облачном API, так и переключаться на локальную модель 14B.
