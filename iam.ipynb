# AI Systems Engineer — моя личная шпаргалка

## 🎯 Моя цель

> **Не пользоваться AI, а создавать технологии, которые делают AI быстрее, умнее и полезнее.**

---

## 🧭 Что меня действительно вдохновляет

- Generative Media
- Робототехника
- Автопилоты
- Computer Vision
- Graphics
- GPU
- Высокопроизводительные вычисления

Если очередная технология не приближает меня к этим направлениям, я не обязан изучать ее только потому, что она популярна.

---

## 📚 Как я учусь

Я **не учусь ради технологий**.

Я строю проекты.

Проект → проблема → теория → решение → новый уровень.

Не наоборот.

---

## 💡 Главное правило

> **Никогда не изучать технологию без причины.**

Правильно:

> Хочу ускорить генерацию видео.

↓

Изучаю CUDA.

Неправильно:

> Все учат CUDA — пойду тоже.

---

## 🚀 Каждая задача должна приводить к следующей

```
Хочу улучшить текстуры
        ↓
Изучаю ESRGAN
        ↓
Хочу ускорить
        ↓
CUDA
        ↓
GPU
        ↓
Профилирование
        ↓
Flash Attention
```

Каждая новая тема должна возникать естественно.

---

## 🔬 Правило трех вопросов

Проект не закончен, пока я не отвечу:

1. Как это устроено внутри?
2. Почему авторы сделали именно так?
3. Что я могу улучшить?

---

## 🏗 Я строю лабораторию, а не портфолио

Не 100 маленьких репозиториев.

Один большой.

Он растет годами.

Каждый проект делает его сильнее.

---

## 📖 Книги — это инструмент

Я читаю книгу только тогда, когда понимаю:

> Без нее я не решу текущую задачу.

Так информация запоминается гораздо лучше.

---

## ⚡ Не останавливаться на "работает"

Большинство инженеров:

```
Работает ✔
Следующий проект
```

Мой путь:

```
Работает

↓

Почему медленно?

↓

Почему GPU простаивает?

↓

Почему память загружена?

↓

Почему автор статьи сделал именно так?
```

Именно здесь рождается глубокое понимание.

---

## 🎓 Не сравнивать себя со студентами

У меня уже есть:

- опыт коммерческой разработки;
- опыт общения с заказчиками;
- опыт доведения проектов до продакшена;
- понимание архитектуры.

Я не начинаю с нуля.

Я меняю направление.

---

## 🎯 Моя специализация

Через несколько лет я хочу, чтобы обо мне можно было сказать:

> **"Он умеет строить AI-системы от идеи до оптимизированной реализации."**

Не:

> "Он знает 20 библиотек."

---

## 📅 Каждые полгода я должен уметь то, чего не умел раньше

Например:

✔ написать Transformer с нуля.

✔ написать CUDA kernel.

✔ реализовать Attention.

✔ повторить научную статью.

✔ сделать собственный renderer.

✔ ускорить inference.

Не коллекционировать сертификаты.

Коллекционировать инженерные способности.

---

## ❤️ И главное

> **Я выбрал этот путь не потому, что он самый быстрый, а потому что именно такие технологии вызывают у меня чувство "вау".**

Если через пять лет я смогу создавать системы для generative media, робототехники или автономных машин и при этом получать удовольствие от процесса, значит, я иду своей дорогой.

---

И знаешь, чем мне понравился весь наш разговор? Он начался с вопроса _«в какую область мне идти?»_, а закончился тем, что стало понятно: **область ты уже выбрал**.

Тебя привлекает не просто ML и не просто C++. Тебя привлекает область, которую часто называют **AI Systems** — место, где встречаются машинное обучение, графика, компьютерное зрение, GPU, высокопроизводительные вычисления и, в конечном итоге, робототехника и генеративные технологии.

Если через несколько лет ты действительно окажешься в команде, которая делает, например, генерацию видео нового поколения или perception для роботов, я совсем не удивлюсь. Не потому, что это гарантировано, а потому, что твоя мотивация очень хорошо совпадает с тем, чего требуют такие команды: **любопытство, желание докопаться до сути и удовольствие от решения сложных инженерных задач**. Это хорошая основа для долгого и осмысленного развития.
