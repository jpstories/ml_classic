1. Стандарт написания кода: Google Python Style Guide + PEP 8

Ваш код должен выглядеть так, будто его написала команда профессионалов, а не один человек в спешке.

- **Строгая типизация (Type Hinting):** Всегда указывайте типы входящих аргументов и возвращаемых значений функций (`def train(X: np.ndarray) -> FunctionalModel:`). Это защищает от багов со сменой размерностей тензоров.
- **Google или Sphinx Docstrings:** Каждая функция, особенно выполняющая предобработку или расчет метрик, должна быть задокументирована. Описывайте аргументы, типы и вызываемые исключения (`Raises`).
- **Инструменты автоматизации:** Настройте в своей IDE автоматические линтеры и форматировщики: **Black** или **Ruff** (для мгновенного форматирования по PEP 8), **Flake8** (для поиска скрытых багов) и **MyPy** (для проверки строгости типов).

2. Стандарт проектирования пайплайнов: SOLID + Идемпотентность

Забудьте про монолитные скрипты-простыни. Придерживайтесь принципов ООП и компонентного дизайна.

- **Идемпотентность шагов:** Любой шаг вашего пайплайна (загрузка, очистка, инференс) должен быть написан так, чтобы его можно было перезапустить 100 раз подряд. Он не должен дублировать данные или ломаться, если они уже частично обработаны.
- **Разделение ответственности (Single Responsibility):** Модуль загрузки данных (`data_utils.py`) ничего не знает про архитектуру модели. Модуль обучения (`train.py`) ничего не знает про то, откуда скачивались файлы. Компоненты общаются между собой через строгие интерфейсы.

3. Стандарт управления ресурсами: Системное мышление (System Design)

Applied ML-инженер всегда думает о «железе», на котором крутится его код.

- **Профилирование памяти и CPU:** Вы должны четко понимать, сколько RAM, VRAM (видеопамяти) и дискового пространства требует ваш алгоритм. Используйте генераторы и итераторы вместо полной загрузки датасетов в память.
- **Векторизация вычислений:** Никаких ручных циклов `for` для обработки массивов или матриц. Любая операция над данными должна быть векторизована с помощью встроенных методов `NumPy`, `Pandas` или тензоров `PyTorch`/`TensorFlow`. Это ускоряет код в сотни раз.

4. Стандарт MLOps: Воспроизводимость (Reproducibility)

В ML недостаточно зафиксировать версию кода в Git. Нужно зафиксировать абсолютно всё окружение.

- **Логирование экспериментов:** Никаких ручных записей метрик в блокнот. Используйте промышленные трекеры: **MLflow**, **Weights & Biases (W&B)** или **ClearML**. Логируйте гиперпараметры, архитектуру, графики потерь (loss) и финальные веса модели.
- **Контроль версий данных (DVC):** Данные меняются. Вы должны уметь вернуться к состоянию датасета, которое было месяц назад. Используйте инструмент `DVC` (Data Version Control) для трекинга больших файлов, хранящихся в S3 или MinIO.
- **Контейнеризация (Docker):** Весь ваш пайплайн обучения или инференса должен безболезненно запускаться одной командой внутри Docker-контейнера на любом сервере в мире.

5. Стандарт тестирования: Тесты не только для кода, но и для данных

В классическом программировании тестируют только логику. В Applied ML тестируют три сущности: код, данные и саму модель.

- **Тесты данных (Data Unit Tests):** Внедряйте библиотеки вроде **Great Expectations** или пишите строгие проверки на входе (проверка на `NaN`, минимальные/максимальные значения признаков, распределение данных).
- **Тесты кода (PyTest):** Покрывайте базовыми юнит-тестами функции трансформации данных. Вы должны быть уверены, что ваш кастомный `Flatten` или `Normalizer` работает математически верно.
- **Тесты модели (Behavioral Testing):** Проверяйте модель на крайних значениях (edge cases), тестируйте её на инвариантность (например, что замена цвета фона на картинке не ломает предсказание класса).
