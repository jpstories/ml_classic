# И самое главное

Я бы ввёл правило, которое редко встречается в учебных планах.

> **Каждый проект должен быть настолько качественным, чтобы через два года вам не было стыдно открыть его снова.**

Это не значит стремиться к идеалу и бесконечно полировать код. Это значит писать так, чтобы проект выглядел как небольшой, но профессионально выполненный open-source репозиторий.

## Мой "золотой стандарт"

Если бы я был вашим ментором, то до завершения каждого проекта я бы проверял следующий чек-лист:

- ✅ Код находится в `src/`, а не в ноутбуке.
- ✅ Notebook используется только для исследований и визуализации.
- ✅ Есть типизация (`type hints`).
- ✅ Есть осмысленные docstrings у публичных функций и классов.
- ✅ Есть проверки размеров тензоров (`shape`) и корректности входных данных там, где это оправдано.
- ✅ Есть логирование вместо хаотичных `print`.
- ✅ Есть базовые тесты (`pytest`) для критичных компонентов.
- ✅ Есть конфигурация через YAML (или аналог), а не захардкоженные параметры.
- ✅ Есть аккуратный `README` с объяснением архитектуры и результатов.
- ✅ Проект автоматически форматируется (`ruff format` или `black`) и проходит проверки линтера (`ruff`) перед каждым коммитом.

Если вы будете придерживаться такого стандарта все два года, то к концу обучения у вас будет не просто набор учебных заданий, а портфолио инженерного уровня. Именно такой подход отличает человека, который **умеет программировать модели**, от человека, который **умеет строить профессиональные ML-системы**.
