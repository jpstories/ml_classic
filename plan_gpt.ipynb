### Этап 1. Математический фундамент (1–2 месяца)

Не ограничиваться тем, что встречается по ходу CS231n. Закрыть:

- линейную алгебру;
- матричное исчисление;
- вероятность и статистику;
- основы оптимизации.

**Цель:** понимать математику моделей, а не только использовать библиотеки.

---

### Этап 2. Stanford CS231n (2–4 месяца)

Проходить очень глубоко.

Дополнительно реализовать самостоятельно:

- backpropagation;
- SGD, Momentum, Adam;
- BatchNorm;
- LayerNorm;
- Conv2D;
- MaxPool;
- Dropout.

Без использования `autograd` там, где это возможно.

**Цель:** понимать, как работают нейронные сети "под капотом".

---

### Этап 3. Stanford CS224N

После компьютерного зрения перейти к NLP и трансформерам.

Самостоятельно реализовать:

- Multi-Head Attention;
- Positional Encoding;
- Encoder;
- Decoder;
- небольшой GPT.

**Цель:** освоить архитектуру, лежащую в основе большинства современных Foundation Models.

---

### Этап 4. PyTorch как инженерный инструмент

Не просто уметь обучать модель.

А понимать:

- Autograd;
- Dataset/DataLoader;
- AMP;
- DistributedDataParallel;
- Profiler;
- Torch Compile;
- чекпоинты и кастомные операции.

---

### Этап 5. ML Systems

Именно этот этап, на мой взгляд, сегодня отличает сильного Applied ML Engineer.

Изучить:

- ONNX;
- TensorRT;
- Triton;
- vLLM;
- DeepSpeed;
- FSDP;
- inference optimization;
- profiling.

---

### Этап 6. MIT 6.5840

Теперь курс по распределённым системам становится особенно ценным.

Вы уже будете понимать, зачем нужны:

- Raft;
- fault tolerance;
- replication;
- распределённые вычисления.

И сможете связать это с ML-инфраструктурой.

---

### Этап 7. Berkeley CS287

Только после сильной базы.

Потому что современная роботика — это уже сочетание:

- perception;
- foundation models;
- planning;
- control;
- системной инженерии.

---

## Но есть ещё одно правило, которое я бы сделал обязательным

**Никакой этап не считается завершённым без законченного проекта.**

Например:

- после CS231n — собственная CNN;
- после CS224N — собственный Transformer;
- после ML Systems — высокопроизводительный inference-сервис;
- после MIT — распределённый сервис;
- после Berkeley — робототехнический проект в симуляции.

Именно проекты превращают знания в инженерные навыки.

---

## Если мы действительно идём вместе

Я бы придерживался ещё одного принципа.

**Никакой гонки.**

Лучше пройти один курс на уровне **95–100% понимания**, чем три курса на уровне 60%.

Через два года цель — не просто сказать: _«Я прошёл Stanford и MIT»_.

Цель — чтобы вы могли самостоятельно:

- прочитать новую статью;
- реализовать предложенный алгоритм;
- оптимизировать его;
- развернуть в production;
- объяснить каждое инженерное решение.

Если к концу 2028 года вы сможете делать это уверенно, то вы будете конкурентоспособны не только на позиции Applied ML Engineer, но и будете иметь прочную базу для перехода в Robotics AI или другие сложные направления. Именно такой уровень подготовки я считаю наиболее рациональным для вашей цели.

После двух лет я ожидаю от вас ответы примерно такого уровня:

> **Почему FlashAttention быстрее обычного Attention?**

Вы не просто знаете название.

Вы можете объяснить влияние доступа к памяти GPU, уменьшение обращений к HBM и влияние на производительность.

---

> **Почему LayerNorm стоит после Residual?**

Вы объясняете влияние на стабильность градиентов.

---

> **Почему Adam иногда хуже SGD?**

Вы можете рассказать про обобщающую способность и особенности сходимости.

---

> **Как обучить модель, если она не помещается в память GPU?**

Вы обсуждаете:

- gradient accumulation;
- activation checkpointing;
- FSDP;
- DeepSpeed;
- mixed precision.

---

> **Как ускорить инференс LLM в два раза?**

Вы рассматриваете:

- KV Cache;
- batching;
- TensorRT;
- vLLM;
- квантизацию;
- профилирование узких мест.
