1. Математика: Топ-5 тем, которые точно спросят

В Сбере любят копать вглубь, чтобы отсеять тех, кто просто вызывает `model.fit()`. Повтори эти концепции:

1. **Градиентный спуск (Gradient Descent):** Как работает backpropagation (обратное распространение ошибки)? Что такое затухание/взрыв градиентов (vanishing/exploding gradients) и как с этим борются (например, через Gradient Clipping или BatchNorm)?
2. **Матричные вычисления:** Как работает операция матричного умножения в контексте полносвязного слоя? Понимание размерностей тензоров — это база для PyTorch.
3. **Метрики классификации и дисбаланс:** Разберись «на пальцах» в геометрическом смысле **ROC-AUC** и **PR-AUC**. Тебя обязательно спросят: _«У нас 99% негативного класса и 1% позитивного (фрод в СберОнлайн). Какую метрику выбрать и почему Accuracy здесь бесполезна?»_.
4. **Регуляризация (L1 и L2):** В чем математическая разница между ними? Почему L1 (Lasso) зануляет веса (делает разреженные матрицы), а L2 (Ridge) просто уменьшает их?
5. **Теория вероятностей и Статистика:** Что такое Центральная Предельная Теорема (ЦПТ)? Как работает p-value и какие ошибки 1-го и 2-го рода бывают при А/Б тестировании фич.

3. ML System Design: Кейс «Кредитный скоринг Сбера»

На этой секции от тебя ждут верхнеуровневый дизайн системы. Вот как должен выглядеть твой ответ по шагам:

```
[ Сырые данные (Транзакции Сбера) ] ➔ [ Расчет фичей в PySpark ] ➔ [ Хранилище Feature Store ]
                                                                             │
[ REST API (FastAPI) ] ➔ [ Сервер инференса (Triton) ] 🦄 [ Модель CatBoost / PyTorch ]
         │
[ Логирование в Kafka ] ➔ [ Мониторинг метрик (Evidently AI / Grafana) ]
```

- **Шаг 1. Формулировка задачи:** Переводим бизнес-цель в ML-метрику. Нам нужно предсказать вероятность дефолта клиента (бинарная классификация). Метрика — ROC-AUC или PR-AUC (так как дефолтов мало).
- **Шаг 2. Данные и Сбор фичей (Твой бэкграунд):** Огромные таблицы транзакций Сбера агрегируются через **PySpark**. Считаем фичи: _«средний чек за 3 месяца»_, _«частота снятия наличных»_. Складываем их в Feature Store (например, Feast).
- **Шаг 3. Моделирование:** Для табличных данных в Сбере до сих пор король — **CatBoost** или **XGBoost**. Если добавляем текстовые данные (описание платежей), можно подключить легкий эмбеддинг-трансформер на **PyTorch**.
- **Шаг 4. Валидация:** Разбиение данных по времени (Time-based split), а не случайный train-test, чтобы избежать заглядывания в будущее (data leakage).
- **Шаг 5. Деплой и Продакшен (MLOps):** Модель упаковывается в Docker. Обертка на **FastAPI** принимает ID пользователя, запрашивает его фичи из Feature Store, отправляет в инференс-движок (Triton) и за миллисекунды отдает ответ: _«одобрить кредит/отказать»_.
- **Шаг 6. Мониторинг:** Логируем предсказания в Kafka. Следим за **Data Drift** (не изменилось ли поведение людей со временем, из-за чего модель начала врать).
